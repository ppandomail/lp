# MiniJSWeb

## Objetivo

Diseñar e implementar un traductor que transforme programas escritos en .minijs Web en código HTML y JavaScript.

## Alcance

- Definir el título de una página.
- Agregar títulos y párrafos.
- Crear campos de texto y numéricos.
- Crear botones.
- Asociar acciones a eventos de clic.
- Mostrar mensajes.
- Escribir contenido en elementos de la página.
- Generar archivo HTML con acciones de JavaScript.

## Especificaciones sintácticas

```text
<Programa> ::= <DeclaracionPagina>
               <ListaInstrucciones>
               <ListaEventos>

<DeclaracionPagina> ::= PAGINA CADENA

<ListaInstrucciones> ::= <Instruccion>
                       | <Instruccion> <ListaInstrucciones>

<Instruccion> ::= <Elemento>

<Elemento> ::= <Titulo>
             | <Texto>
             | <Contenedor>
             | <Campo>
             | <Boton>

<Titulo> ::= TITULO CADENA

<Texto> ::= TEXTO CADENA
          | TEXTO CADENA <ListaAtributos>

<Contenedor> ::= CONTENEDOR <ListaElementos> FIN_CONTENEDOR
               | CONTENEDOR <ListaAtributos> <ListaElementos> FIN_CONTENEDOR

<ListaElementos> ::= <Elemento>
                   | <Elemento> <ListaElementos>

<Campo> ::= CAMPO TIPO_CAMPO
          | CAMPO TIPO_CAMPO <ListaAtributos>

<Boton> ::= BOTON CADENA
          | BOTON CADENA <ListaAtributos>

<ListaAtributos> ::= ID CADENA
                   | CLASE CADENA
                   | ID CADENA CLASE CADENA

<ListaEventos> ::= <Evento>
                 | <Evento> <ListaEventos>

<Evento> ::= AL_HACER_CLICK_EN IDENTIFICADOR
             <ListaAcciones>
             FIN_EVENTO

<ListaAcciones> ::= <Accion>
                  | <Accion> <ListaAcciones>

<Accion> ::= MOSTRAR CADENA
           | ESCRIBIR CADENA EN IDENTIFICADOR
           | AGREGAR CLASE CADENA A IDENTIFICADOR
           | QUITAR CLASE CADENA DE IDENTIFICADOR
```

## Especificaciones Semanticas

| *Regla* | *Acciones Semánticas* |
|---|---|
| `<Programa> ::= <DeclaracionPagina> <ListaInstrucciones> <ListaEventos>` | Generar el documento HTML y JavaScript completo. |
| `<ListaInstrucciones> ::= <Instruccion>` | Generar el código HTML correspondiente a la instrucción. |
| `<ListaInstrucciones> ::= <Instruccion> <ListaInstrucciones>` | 	Concatenar el código HTML de ambas partes, respetando el orden original. |
| `<Instruccion> ::= <Elemento>` | Procesar elemento. |
| `<Elemento> ::= <Titulo>` | Procesar título. |
| `<Elemento> ::= <Texto>` | Procesar texto. |
| `<Elemento> ::= <Contenedor>` | Procesar contenedor. |
| `<Elemento> ::= <Campo>` | Procesar campo. |
| `<Elemento> ::= <Boton>` | Procesar botón. |
| `<DeclaracionPagina> ::= PAGINA CADENA` | Establece el título dentro de <head>: `Trad: <title>CADENA</title>`. |
| `<Titulo> ::= TITULO CADENA` | `Trad: <h1>CADENA</h1>`. |
| `<Texto> ::= TEXTO CADENA` | `Trad: <p>CADENA</p>`. |
| `<Texto> ::= TEXTO CADENA <ListaAtributos>` | `Trad: <p ATRIBUTOS>CADENA</p>`. |
| `<Contenedor> ::= CONTENEDOR <ListaElementos> FIN_CONTENEDOR` | `Trad: <div>ELEMENTOS</div>`. |
| `<Contenedor> ::= CONTENEDOR <ListaAtributos> <ListaElementos> FIN_CONTENEDOR` | `Trad: <div ATRIBUTOS>ELEMENTOS</div>`. |
| `<ListaElementos> ::= <Elemento>` | Generar el código HTML del elemento. |
| `<ListaElementos> ::= <Elemento> <ListaElementos>` | Concatenar el código HTML de los elementos, respetando el orden original. |
| `<Campo> ::= CAMPO TIPO_CAMPO` | `Trad: <input type="TIPO_CAMPO">`. |
| `<Campo> ::= CAMPO TIPO_CAMPO <ListaAtributos>` | `Trad: <input type="TIPO_CAMPO" ATRIBUTOS>`. |
| `<Boton> ::= BOTON CADENA` | `Trad: <button>CADENA</button>`. |
| `<Boton> ::= BOTON CADENA <ListaAtributos>` | `Trad: <button ATRIBUTOS>CADENA</button>`. |
| `<ListaAtributos> ::= ID CADENA` | `Trad: id="CADENA"`. Registrar el ID. |
| `<ListaAtributos> ::= CLASE CADENA` | `Trad: class="CADENA"`. |
| `<ListaAtributos> ::= ID CADENA CLASE CADENA` | `Trad: id="CADENA" class="CADENA"`. Registrar el ID. |
| `<ListaEventos> ::= <Evento>` | Generar el código JavaScript correspondiente al evento. |
| `<ListaEventos> ::= <Evento> <ListaEventos>` | Concatenar las definiciones de eventos. |
| `<Evento> ::= AL_HACER_CLICK_EN IDENTIFICADOR <ListaAcciones> FIN_EVENTO` | Generar un evento JavaScript mediante addEventListener("click", function() { ACCIONES });, asociado al elemento identificado. |
| `<ListaAcciones> ::= <Accion>` | Generar el código JavaScript correspondiente a la acción. |
| `<ListaAcciones> ::= <Accion> <ListaAcciones>` | Concatenar las acciones JavaScript, respetando el orden original. |
| `<Accion> ::= MOSTRAR CADENA` | Generar alert(CADENA); para mostrar un mensaje. |
| `<Accion> ::= ESCRIBIR CADENA EN IDENTIFICADOR` | Generar una asignación de contenido al elemento indicado. Utilizar .value para campos de entrada y .textContent para los demás elementos. |
| `<Accion> ::= AGREGAR CLASE CADENA A IDENTIFICADOR` | Generar document.getElementById("IDENTIFICADOR").classList.add(CADENA);. |
| `<Accion> ::= QUITAR CLASE CADENA DE IDENTIFICADOR` | Generar document.getElementById("IDENTIFICADOR").classList.remove(CADENA);. |

````python
%%file tp_trad/lexer_minijs.py

import ply.lex as lex

errores_lexicos = [False]

tokens = (
    'PAGINA',
    'TITULO',
    'TEXTO',
    'CONTENEDOR',
    'FIN_CONTENEDOR',
    'CAMPO',
    'BOTON',
    'ID',
    'CLASE',
    'AL_HACER_CLICK_EN',
    'FIN_EVENTO',
    'MOSTRAR',
    'ESCRIBIR',
    'AGREGAR',
    'QUITAR',
    'A',
    'DE',
    'EN',
    'CADENA',
    'IDENTIFICADOR',
    'TIPO_CAMPO'
)

def t_PAGINA(t):
    r'PAGINA\b'
    return t

def t_TITULO(t):
    r'TITULO\b'
    return t

def t_TEXTO(t):
    r'TEXTO\b'
    return t

def t_CONTENEDOR(t):
    r'CONTENEDOR\b'
    return t

def t_FIN_CONTENEDOR(t):
    r'FIN_CONTENEDOR\b'
    return t

def t_CAMPO(t):
    r'CAMPO\b'
    return t

def t_BOTON(t):
    r'BOTON\b'
    return t

def t_ID(t):
    r'ID\b'
    return t

def t_CLASE(t):
    r'CLASE\b'
    return t

def t_AL_HACER_CLICK_EN(t):
    r'AL_HACER_CLICK_EN\b'
    return t

def t_FIN_EVENTO(t):
    r'FIN_EVENTO\b'
    return t

def t_MOSTRAR(t):
    r'MOSTRAR\b'
    return t

def t_ESCRIBIR(t):
    r'ESCRIBIR\b'
    return t

def t_AGREGAR(t):
    r'AGREGAR\b'
    return t

def t_QUITAR(t):
    r'QUITAR\b'
    return t

def t_A(t):
    r'A\b'
    return t

def t_DE(t):
    r'DE\b'
    return t

def t_EN(t):
    r'EN\b'
    return t

def t_TIPO_CAMPO(t):
    r'text\b|number\b|password\b'
    return t

def t_CADENA(t):
    r'"([^"\\]|\\.)*"|\'([^\'\\]|\\.)*\''
    t.value = t.value[1:-1]
    return t

def t_IDENTIFICADOR(t):
    r'[a-zA-Z_][a-zA-Z0-9_-]*'
    return t

t_ignore = ' \t'

def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    print(f"Error léxico: carácter ilegal '{t.value[0]}'")
    errores_lexicos[0] = True
    t.lexer.skip(1)


lexer = lex.lex()
````

````python
%%file tp_trad/generar_html.py

import ollama


def generar_tag_html(tag, contenido, atributos):

    respuesta = ollama.chat(
        model='llama3.2',
        messages=[
            {
                'role': 'user',
                'content': f'''
                Genera únicamente una etiqueta HTML válida.

                Etiqueta: {tag}
                Atributos: {atributos}
                Contenido: {contenido}

                Reglas:
                Reglas obligatorias:
                - No agregues explicaciones.
                - No agregues markdown.
                - No inventes atributos.
                - No modifiques los atributos recibidos.
                - No modifiques el contenido recibido.
                - No agregues identificadores ni clases adicionales.
                - Si no hay atributos, no agregues ninguno.
                - Para input, utiliza una etiqueta sin cierre.
                - Para las demás etiquetas, utiliza apertura y cierre.
                - Devuelve una sola etiqueta HTML.
                '''
            }
        ]
    )

    return respuesta.message.content.strip()
````

````python
%%file tp_trad/parser_minijs.py

import ply.yacc as yacc
import json
from lexer_minijs import tokens
from generar_html import generar_tag_html


html = ['<!DOCTYPE html>', '<html>', '<head>', '<meta charset="UTF-8">']
ids_registrados = set()
errores_semanticos = [False]

def p_programa(p):
    '''
    programa : declaracion_pagina lista_instrucciones lista_eventos
    '''

    html.append('</head>')
    html.append('<body>')
    html.append(p[2])
    html.append('<script>')
    html.append(p[3])
    html.append('</script>')
    html.append('</body>')
    html.append('</html>')

def p_declaracion_pagina(p):
    'declaracion_pagina : PAGINA CADENA'

    html.append('<title>' + p[2] + '</title>')


def p_lista_instrucciones(p):
    '''
    lista_instrucciones : instruccion
                        | lista_instrucciones instruccion
    '''

    if len(p) == 2:
        p[0] = p[1]
    else:
        p[0] = p[1] + p[2]


def p_instruccion(p):
    'instruccion : elemento'

    p[0] = p[1]


def p_elemento(p):
    '''
    elemento : titulo
             | texto
             | contenedor
             | campo
             | boton
    '''

    p[0] = p[1]



def p_titulo(p):
    'titulo : TITULO CADENA'
    p[0] = generar_tag_html('h1', p[2], '') + '\n'


def p_texto(p):
    '''
    texto : TEXTO CADENA
          | TEXTO CADENA lista_atributos
    '''
    if len(p) == 3:
        atributos = ''
    else:
        atributos = p[3]

    p[0] = generar_tag_html('p', p[2], atributos) + '\n'


def p_contenedor(p):
    '''
    contenedor : CONTENEDOR lista_elementos FIN_CONTENEDOR
               | CONTENEDOR lista_atributos lista_elementos FIN_CONTENEDOR
    '''
    if len(p) == 4:
        p[0] = '<div>\n' + p[2] + '</div>\n'
    else:
        p[0] = '<div ' + p[2] + '>\n' + p[3] + '</div>\n'



def p_lista_elementos(p):
    '''
    lista_elementos : elemento
                    | lista_elementos elemento
    '''

    if len(p) == 2:
        p[0] = p[1]
    else:
        p[0] = p[1] + p[2]


def p_campo(p):
    '''
    campo : CAMPO TIPO_CAMPO
          | CAMPO TIPO_CAMPO lista_atributos
    '''
    if len(p) == 3:
        atributos = 'type="' + p[2] + '"'
    else:
        atributos = 'type="' + p[2] + '" ' + p[3]

    p[0] = generar_tag_html('input', '', atributos) + '\n'


def p_boton(p):
    '''
    boton : BOTON CADENA
          | BOTON CADENA lista_atributos
    '''
    if len(p) == 3:
        atributos = ''
    else:
        atributos = p[3]

    p[0] = generar_tag_html('button', p[2], atributos) + '\n'


def p_lista_atributos_id(p):
    'lista_atributos : ID CADENA'
    ids_registrados.add(p[2])
    p[0] = 'id="' + p[2] + '"'


def p_lista_atributos_clase(p):
    'lista_atributos : CLASE CADENA'
    p[0] = 'class="' + p[2] + '"'

def p_lista_atributos_id_clase(p):
    'lista_atributos : ID CADENA CLASE CADENA'
    ids_registrados.add(p[2])
    p[0] = 'id="' + p[2] + '" class="' + p[4] + '"'

def p_lista_eventos(p):
    '''
    lista_eventos : evento
                  | lista_eventos evento
    '''

    if len(p) == 2:
        p[0] = p[1]
    else:
        p[0] = p[1] + p[2]

def p_evento(p):
    'evento : AL_HACER_CLICK_EN IDENTIFICADOR lista_acciones FIN_EVENTO'
    
    if p[2] not in ids_registrados:
        print(f"Error semántico en línea {p.lineno(2)}: El elemento con ID '{p[2]}' no existe.")
        errores_semanticos[0] = True
        p[0] = ""
        return

    p[0] = (
        'document.getElementById(' + json.dumps(p[2]) +
        ').addEventListener("click", function() {\n' +
        p[3] +
        '});\n'
    )
    
def p_lista_acciones(p):
    '''
    lista_acciones : accion
                   | lista_acciones accion
    '''

    if len(p) == 2:
        p[0] = p[1]
    else:
        p[0] = p[1] + p[2]



def p_accion_mostrar(p):
    'accion : MOSTRAR CADENA'

    p[0] = 'alert(' + json.dumps(p[2]) + ');\n'


def p_accion_escribir(p):
    'accion : ESCRIBIR CADENA EN IDENTIFICADOR'

    p[0] = (
        '(() => {\n'
        '    const elemento = document.getElementById('
        + json.dumps(p[4]) + ');\n'
        '    if (elemento) {\n'
        '        if (elemento.tagName === "INPUT") {\n'
        '            elemento.value = ' + json.dumps(p[2]) + ';\n'
        '        } else {\n'
        '            elemento.textContent = ' + json.dumps(p[2]) + ';\n'
        '        }\n'
        '    }\n'
        '})();\n'
    )

def p_accion_agregar(p):
    'accion : AGREGAR CLASE CADENA A IDENTIFICADOR'

    p[0] = (
        'document.getElementById(' + json.dumps(p[5]) +
        ').classList.add(' + json.dumps(p[3]) + ');\n'
    )


def p_accion_quitar(p):
    'accion : QUITAR CLASE CADENA DE IDENTIFICADOR'

    p[0] = (
        'document.getElementById(' + json.dumps(p[5]) +
        ').classList.remove(' + json.dumps(p[3]) + ');\n'
    )



def p_error(p):
    if p:
        print(f"Error sintáctico en línea {p.lineno}: "
              f"Token inesperado '{p.value}' ({p.type})")
    else:
        print("Error sintáctico: fin inesperado del archivo.")

    html.clear()
    html.append('Error sintáctico')

parser = yacc.yacc()
````

````python
%%file tp_trad/ejemplo.minijs

PAGINA "Sistema de Registro"

TITULO "Registro de usuarios"

CONTENEDOR ID "contenedor" CLASE "formulario"

    TEXTO "Bienvenido"

    TEXTO "Ingrese su nombre:"

    CAMPO text ID "nombreUsuario" CLASE "campo"

    CONTENEDOR ID "acciones"

        BOTON "Registrar" ID "btnRegistrar" CLASE "boton"

        TEXTO "Esperando registro..." ID "mensaje"

    FIN_CONTENEDOR

FIN_CONTENEDOR

AL_HACER_CLICK_EN btnRegistrar

    ESCRIBIR "Registro realizado correctamente" EN mensaje

    AGREGAR CLASE "exito" A mensaje

    MOSTRAR "Usuario registrado"

FIN_EVENTO
````

````python
%%file tp_trad/main.py

import os
import sys

from lexer_minijs import lexer, errores_lexicos
from parser_minijs import parser, html, errores_semanticos 

def main():

    carpeta = os.path.dirname(os.path.abspath(__file__))

    if len(sys.argv) > 1:
        nombre_archivo = os.path.abspath(sys.argv[1])
    else:
        nombre_archivo = os.path.join(carpeta, 'ejemplo.minijs')

    print("Ruta del archivo:", nombre_archivo)
    print("¿Existe el archivo?", os.path.exists(nombre_archivo))

    with open(nombre_archivo, 'r', encoding='utf-8') as entrada:
        codigo = entrada.read()

    html.clear()
    html.extend([
        '<!DOCTYPE html>',
        '<html>',
        '<head>',
        '<meta charset="UTF-8">'
    ])

    errores_lexicos[0] = False
    errores_semanticos[0] = False

    parser.parse(codigo, lexer=lexer)

    if errores_semanticos[0]:
        print('Se encontraron errores semánticos. No se generará el archivo.')
        return

    if errores_lexicos[0]:
        print('Se encontraron errores léxicos.')
        return

    if 'Error sintáctico' in html:
        print('Se encontraron errores sintácticos.')
        return

    nombre_salida = os.path.splitext(nombre_archivo)[0] + '.html'

    with open(nombre_salida, 'w', encoding='utf-8') as archivo:
        archivo.write('\n'.join(html))

    print('Archivo HTML generado correctamente:', nombre_salida)

if __name__ == '__main__':
    main()
````

````python
%%file tp_trad/tests.py

import subprocess
import sys
from pathlib import Path


def test_caso_feliz():

    archivo = Path("tp_trad/ejemplo.minijs")

    resultado = subprocess.run(
        [sys.executable, "tp_trad/main.py", str(archivo)],
        capture_output=True,
        text=True,
        encoding="utf-8" 
    )
    
    assert resultado.returncode == 0

    html_generado = Path("tp_trad/ejemplo.html")

    assert html_generado.exists()

    actual = html_generado.read_text(encoding="utf-8")

    assert "Registro de usuarios" in actual
    assert "btnRegistrar" in actual
    assert 'addEventListener("click"' in actual
    assert 'alert("Usuario registrado")' in actual


def test_caso_no_feliz():

    archivo = Path("tp_trad/ejemplo_err.minijs")
    html_generado = Path("tp_trad/ejemplo_err.html")

    if html_generado.exists():
        html_generado.unlink()

    resultado = subprocess.run(
        [sys.executable, "tp_trad/main.py", str(archivo)],
        capture_output=True,
        text=True,
        encoding="utf-8" 
    )

    assert resultado.returncode == 0
    assert "errores sint" in resultado.stdout
    assert not html_generado.exists()

def test_contenedores():

    archivo = Path("tp_trad/contenedores.minijs")

    resultado = subprocess.run(
        [sys.executable, "tp_trad/main.py", str(archivo)],
        capture_output=True,
        text=True,
        encoding="utf-8"
    )

    assert resultado.returncode == 0

    html_generado = Path("tp_trad/contenedores.html")

    assert html_generado.exists()

    actual = html_generado.read_text(encoding="utf-8")

    assert "principal" in actual
    assert "secundario" in actual

def test_error_lexico():

    archivo = Path("tp_trad/error_lexico_1.minijs")
    html_generado = Path("tp_trad/error_lexico_1.html")

    if html_generado.exists():
        html_generado.unlink()

    resultado = subprocess.run(
        [sys.executable, "tp_trad/main.py", str(archivo)],
        capture_output=True,
        text=True,
        encoding="utf-8" 
    )

    assert resultado.returncode == 0
    assert "léxico" in resultado.stdout.lower()
    assert not html_generado.exists()

def test_error_semantico():

        archivo = Path("tp_trad/ejemplo_err_semant.minijs")
        html_generado = Path("tp_trad/ejemplo_err_semant.html")

        if html_generado.exists():
            html_generado.unlink()

        resultado = subprocess.run(
            [sys.executable, "tp_trad/main.py", str(archivo)],
            capture_output=True,
            text=True,
            encoding="utf-8" 
        )

        assert resultado.returncode == 0
        assert "semántico" in resultado.stdout.lower()
        assert not html_generado.exists()
````
