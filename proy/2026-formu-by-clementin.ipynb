# Formu

## Objetivo

Transformar programas escritos en el lenguaje **Formu** en formularios web en código **HTML**, con sus campos, etiquetas y validaciones.

## Ejemplos

Dos programas válidos según la GIC: uno simple y otro que usa todos los tipos y modificadores. Para cada uno se muestra el programa fuente, la traducción a HTML que genera el traductor y el formulario ya ejecutado. Al apretar el botón, si los campos obligatorios están completos aparece **Operación realizada** y los datos no se envían a otro lado; si falta algo, el navegador marca el campo y no aparece el mensaje.

## Alcance

* **Encabezado del formulario**: `formulario "Título" enviar "/ruta"` define el título de la página (`<title>` y `<h1>`) y la dirección a la que se envían los datos (`<form action="/ruta" method="post">`)
* **Campos**: `campo id tipo "Etiqueta"` agrega un campo al formulario. El `id` se usa como `id` y `name` del HTML y la etiqueta como texto del `<label>`
* **7 tipos de campo**:
  * `texto` → caja de texto (`<input type="text">`)
  * `numero` → número (`<input type="number">`)
  * `correo` → correo electrónico, el navegador valida que tenga `@` (`<input type="email">`)
  * `clave` → contraseña oculta (`<input type="password">`)
  * `fecha` → selector de fecha (`<input type="date">`)
  * `casilla` → casilla de verificación (`<input type="checkbox">`)
  * `opciones` → lista desplegable (`<select>` con un `<option>` por valor)
* **Modificadores opcionales** que se escriben después de la etiqueta, en cualquier orden:
  * `requerido` → el campo es obligatorio (`required`)
  * `min N` y `max N` → valor mínimo y máximo de un campo `numero` (`min="N"`, `max="N"`)
  * `["A", "B", "C"]` → valores de un campo `opciones`
* **Botón de envío**: `boton "Texto"` cierra el formulario (`<button type="submit">`)
* **Confirmación al enviar**: al apretar el botón, solo si el navegador validó todos los campos (obligatorios, mínimos y máximos, formato de correo), aparece el mensaje **Operación realizada** debajo del botón y el formulario no se envía a otra página
* **Comentarios**: todo lo que sigue a `#` hasta el fin de la línea se ignora
* **Placeholders con un agente LLM**: los campos texto, numero, correo y clave llevan un texto de ejemplo que genera un agente (`ag_generar_placeholder.py`) pidiéndoselo a **Ollama** con el modelo `llama3.2`. Si Ollama no está disponible (por ejemplo en jupyter.org), el agente usa un texto fijo según el tipo (`Escribí tu nombre completo`, `nombre@ejemplo.com`)
* **Detección de errores** con número de línea:
  * Léxicos: caracteres que no pertenecen al lenguaje (por ejemplo `@`)
  * Sintácticos: instrucciones mal formadas o incompletas (falta el id, falta el botón)
  * Semánticos: campo declarado dos veces, `min`/`max` en un campo que no es `numero`, `min` mayor que `max`, campo `opciones` sin lista, lista en un campo que no es `opciones`, modificador repetido y destino que no empieza con `/` o `http`
* **HTML legible**: el código generado sale con sangría de 2 espacios por nivel, una etiqueta por línea, como se escribe HTML a mano
* **Salida**: si el programa es válido se genera el archivo `<programa>.formu.html` con la página completa; si hay errores, el archivo contiene solo los mensajes de error
* **Fuera de alcance**: estilos CSS, JavaScript, campos agrupados o anidados, y más de un formulario por archivo

## Especificaciones léxicas

| Lexema | Token | Patrón |
| -- | -- | -- |
| formulario | FORMULARIO | formulario |
| enviar | ENVIAR | enviar |
| campo | CAMPO | campo |
| boton | BOTON | boton |
| requerido | REQUERIDO | requerido |
| min | MIN | min |
| max | MAX | max |
| texto | TIPO | texto |
| numero | TIPO | numero |
| correo | TIPO | correo |
| clave | TIPO | clave |
| fecha | TIPO | fecha |
| casilla | TIPO | casilla |
| opciones | TIPO | opciones |
| nombre, email, fecha_nac | ID | (L \| _) (L \| D \| _)* |
| "Nombre completo", "/login" | CADENA | " C* " |
| 17, 99 | NUMERO | D+ |
| [ | CORCHETE_A | [ |
| ] | CORCHETE_C | ] |
| , | COMA | , |

Referencias: **L** = letra (a…z, A…Z), **D** = dígito (0…9), **C** = cualquier carácter excepto `"` y salto de línea.

Elementos que el scanner reconoce pero no envía al parser:

| Lexema | Patrón |
| -- | -- |
| # comentario | # C* |
| espacio, tabulación | ( ␣ \| \t )+ |
| salto de línea (se usa para contar líneas) | \n+ |

Las palabras reservadas cumplen el mismo patrón que ID. El scanner reconoce primero el lexema como ID y luego lo busca en la tabla de palabras reservadas: si lo encuentra, le asigna el token de la tabla (por ejemplo `campo` → CAMPO, `texto` → TIPO).

## Especificaciones sintácticas

```
<Programa> ::= <Encabezado> <Campos> <Boton>
<Encabezado> ::= FORMULARIO CADENA ENVIAR CADENA
<Campos> ::= <Campos> <Campo>
<Campos> ::= <Campo>
<Campo> ::= CAMPO ID TIPO CADENA <Modificadores>
<Modificadores> ::= <Modificadores> <Modificador>
<Modificadores> ::= λ
<Modificador> ::= REQUERIDO
<Modificador> ::= MIN NUMERO
<Modificador> ::= MAX NUMERO
<Modificador> ::= CORCHETE_A <Valores> CORCHETE_C
<Valores> ::= <Valores> COMA CADENA
<Valores> ::= CADENA
<Boton> ::= BOTON CADENA
```

* Símbolos no terminales: `<Programa>`, `<Encabezado>`, `<Campos>`, `<Campo>`, `<Modificadores>`, `<Modificador>`, `<Valores>`, `<Boton>`
* Símbolos terminales: los tokens de la grilla léxica
* Axioma: `<Programa>`
* λ representa la cadena vacía: un campo puede no tener modificadores

## Especificaciones semánticas

| Regla | Acciones Semánticas |
| -- | -- |
| \<Programa> ::= \<Encabezado> \<Campos> \<Boton> | Un formulario completo. Trad: `</form>`. Si no hubo errores, envuelve todo en `<html><head><title>TÍTULO</title></head><body> … </body></html>` |
| \<Encabezado> ::= FORMULARIO CADENA₁ ENVIAR CADENA₂ | Guarda CADENA₁ como título. Verifica que CADENA₂ empiece con `/` o `http`; si no, error semántico "destino inválido". Trad: `<h1>CADENA₁</h1>` `<form action="CADENA₂" method="post" onsubmit="…">`. El `onsubmit` cancela el envío a otra página (`event.preventDefault()`) y escribe "Operación realizada" en el `<output>` del formulario |
| \<Campos> ::= \<Campos> \<Campo> | Muchos campos |
| \<Campos> ::= \<Campo> | Un campo |
| \<Campo> ::= CAMPO ID TIPO CADENA \<Modificadores> | Si ID ya está en la tabla de símbolos: error "campo ya declarado"; si no, agrega ID → TIPO. Si hay min o max y TIPO no es `numero`: error. Si min > max: error. Si TIPO es `opciones` y no hay lista: error. Si hay lista y TIPO no es `opciones`: error. Para los tipos texto, numero, correo y clave le pide el placeholder al agente LLM `generar_placeholder(TIPO, CADENA)` (Ollama, o la tabla fija si Ollama no está). Trad: `<label for="ID">CADENA</label>` `<input type="T" id="ID" name="ID" placeholder="…" [min="N"] [max="N"] [required]>`; para `opciones`: `<select id="ID" name="ID">` + un `<option>` por valor + `</select>` |
| \<Modificadores> ::= \<Modificadores> \<Modificador> | Agrega el modificador al conjunto de modificadores del campo. Si ya estaba: error "modificador repetido" |
| \<Modificadores> ::= λ | Conjunto vacío de modificadores |
| \<Modificador> ::= REQUERIDO | Campo obligatorio. Trad: `required` |
| \<Modificador> ::= MIN NUMERO | Valor mínimo = NUMERO. Trad: `min="NUMERO"` |
| \<Modificador> ::= MAX NUMERO | Valor máximo = NUMERO. Trad: `max="NUMERO"` |
| \<Modificador> ::= CORCHETE_A \<Valores> CORCHETE_C | Lista de valores del campo = Valores |
| \<Valores> ::= \<Valores> COMA CADENA | Agrega CADENA a la lista. Trad: `<option>CADENA</option>` |
| \<Valores> ::= CADENA | Lista con un valor. Trad: `<option>CADENA</option>` |
| \<Boton> ::= BOTON CADENA | Trad: `<button type="submit">CADENA</button>` `<output></output>`. El `<output>` empieza vacío; al apretar el botón, el navegador primero valida los campos (`required`, `min`, `max`, `email`) y solo si todo es válido se ejecuta el `onsubmit`, que muestra **Operación realizada** sin enviar los datos a otro lado |

Traducción de TIPO: texto → `text`, numero → `number`, correo → `email`, clave → `password`, fecha → `date`, casilla → `checkbox`, opciones → `<select>`.

**Formato del HTML generado**: cada acción semántica escribe su línea con `emitir(nivel, línea)`, que agrega 2 espacios de sangría por cada nivel de anidamiento (`<html>` 0, `<head>`/`<body>` 1, `<h1>`/`<form>` 2, `<div>`/`<button>`/`<output>` 3, `<label>`/`<input>`/`<select>` 4, `<option>` 5). Así el archivo generado queda indentado como se escribe HTML a mano: una etiqueta por línea y cada hijo un nivel más adentro que su padre. La sangría no cambia cómo se ve la página (el navegador la ignora), solo la hace legible.

````python
%%file formu/lexer_formu.py
import ply.lex as lex

reservadas = {
    'formulario': 'FORMULARIO',
    'enviar':     'ENVIAR',
    'campo':      'CAMPO',
    'boton':      'BOTON',
    'requerido':  'REQUERIDO',
    'min':        'MIN',
    'max':        'MAX',
    'texto':      'TIPO',
    'numero':     'TIPO',
    'correo':     'TIPO',
    'clave':      'TIPO',
    'fecha':      'TIPO',
    'casilla':    'TIPO',
    'opciones':   'TIPO',
}

tokens = (
    'FORMULARIO', 'ENVIAR', 'CAMPO', 'BOTON', 'REQUERIDO', 'MIN', 'MAX',
    'TIPO', 'ID', 'CADENA', 'NUMERO', 'CORCHETE_A', 'CORCHETE_C', 'COMA',
)

errores_lexicos = []

t_CORCHETE_A = r'\['
t_CORCHETE_C = r'\]'
t_COMA       = r','
t_ignore     = ' \t'
t_ignore_COMENTARIO = r'\#.*'

def t_CADENA(t):
    r'"[^"\n]*"'
    t.value = t.value[1:-1]
    return t

def t_NUMERO(t):
    r'\d+'
    t.value = int(t.value)
    return t

def t_ID(t):
    r'[a-zA-Z_][a-zA-Z0-9_]*'
    t.type = reservadas.get(t.value, 'ID')
    return t

def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    errores_lexicos.append(f"Error léxico: carácter ilegal '{t.value[0]}' en línea {t.lineno}")
    t.lexer.skip(1)

lexer = lex.lex()
````

````python
%%file formu/ag_generar_placeholder.py
import os
import sys

# Agente LLM: le pide a Ollama (modelo llama3.2) el texto de ejemplo
# (placeholder) de un campo. Si Ollama no está disponible, usa una tabla fija.

MODELO = 'llama3.2'

# textos por defecto según el tipo de campo
PLACEHOLDERS = {'texto':  'Escribí tu {e}',
                'numero': 'Ingresá un número',
                'correo': 'nombre@ejemplo.com',
                'clave':  'Mínimo 8 caracteres'}

ultimo_origen = {'valor': ''}   # 'Ollama' o 'tabla fija': lo muestra el notebook


def placeholder_por_defecto(tipo, etiqueta):
    ultimo_origen['valor'] = 'tabla fija'
    return PLACEHOLDERS[tipo].format(e=etiqueta.lower())


def llm_disponible():
    # FORMU_LLM=0 apaga el agente (lo usan los tests para tener una salida fija).
    # En jupyter.org (Pyodide, sys.platform 'emscripten') Ollama no puede correr.
    if os.environ.get('FORMU_LLM', '1') == '0' or sys.platform == 'emscripten':
        return False
    try:
        import ollama  # noqa: F401
        return True
    except ImportError:
        return False


def generar_placeholder(tipo, etiqueta):
    if not llm_disponible():
        return placeholder_por_defecto(tipo, etiqueta)
    try:
        import ollama
        respuesta = ollama.Client(timeout=60).chat(
            model=MODELO,
            messages=[{
                'role': 'user',
                'content': f'''
                Devuelve únicamente el texto, en castellano.
                Sin explicaciones. Sin comillas. Máximo 6 palabras.
                Escribe un texto de ejemplo (placeholder) para un campo
                de formulario web de tipo {tipo} cuya etiqueta es: {etiqueta}
                '''
            }]
        )
        texto = respuesta['message']['content'].strip().strip('"').splitlines()[0].strip()
        if 0 < len(texto) <= 60:
            ultimo_origen['valor'] = 'Ollama'
            return texto
        return placeholder_por_defecto(tipo, etiqueta)   # respuesta inválida
    except Exception:
        # Ollama no está abierto o no tiene el modelo descargado
        return placeholder_por_defecto(tipo, etiqueta)
````

````python
%%file formu/parser_formu.py
import ply.yacc as yacc
from html import escape
from lexer_formu import tokens, lexer, errores_lexicos
from ag_generar_placeholder import generar_placeholder   # agente LLM (Ollama)

html = []
errores = []
tabla_campos = {}     # tabla de símbolos: id -> tipo
info = {'titulo': ''}

TIPOS_HTML = {'texto': 'text', 'numero': 'number', 'correo': 'email',
              'clave': 'password', 'fecha': 'date', 'casilla': 'checkbox'}

# al apretar el botón: si el navegador validó todo, no se envía a otro lado
# y se muestra el mensaje de confirmación dentro del mismo formulario
AL_ENVIAR = ("event.preventDefault(); "
             "this.querySelector('output').textContent = 'Operación realizada'")

# sangría del HTML: 2 espacios por cada nivel de anidamiento
SANGRIA = '  '

def emitir(nivel, linea):
    html.append(SANGRIA * nivel + linea)

def error_semantico(msg, linea):
    errores.append(f"Error semántico: {msg} en línea {linea}")

# ---------- Gramática y acciones semánticas ----------

def p_programa(p):
    'programa : encabezado campos boton'
    emitir(2, '</form>')

def p_encabezado(p):
    'encabezado : FORMULARIO CADENA ENVIAR CADENA'
    if not (p[4].startswith('/') or p[4].startswith('http')):
        error_semantico(f"el destino '{p[4]}' debe empezar con / o http", p.lineno(3))
    info['titulo'] = escape(p[2])
    emitir(2, f'<h1>{escape(p[2])}</h1>')
    emitir(2, f'<form action="{escape(p[4])}" method="post" onsubmit="{AL_ENVIAR}">')

def p_campos(p):
    '''campos : campos campo
              | campo'''

def p_campo(p):
    'campo : CAMPO ID TIPO CADENA modificadores'
    ident, tipo, etiqueta, mods, linea = p[2], p[3], p[4], p[5], p.lineno(1)

    if ident in tabla_campos:
        error_semantico(f"el campo '{ident}' ya fue declarado", linea)
    tabla_campos[ident] = tipo

    if ('min' in mods or 'max' in mods) and tipo != 'numero':
        error_semantico(f"min/max solo se permiten en campos numero ('{ident}' es {tipo})", linea)
    if 'min' in mods and 'max' in mods and mods['min'] > mods['max']:
        error_semantico(f"en '{ident}' min ({mods['min']}) es mayor que max ({mods['max']})", linea)
    if tipo == 'opciones' and 'lista' not in mods:
        error_semantico(f"el campo opciones '{ident}' necesita una lista [...]", linea)
    if tipo != 'opciones' and 'lista' in mods:
        error_semantico(f"solo los campos opciones llevan lista ('{ident}' es {tipo})", linea)

    req = ' required' if mods.get('requerido') else ''
    emitir(3, '<div>')
    if tipo == 'opciones':
        emitir(4, f'<label for="{ident}">{escape(etiqueta)}</label>')
        emitir(4, f'<select id="{ident}" name="{ident}"{req}>')
        for valor in mods.get('lista', []):
            emitir(5, f'<option>{escape(valor)}</option>')
        emitir(4, '</select>')
    elif tipo == 'casilla':
        emitir(4, f'<input type="checkbox" id="{ident}" name="{ident}"{req}>')
        emitir(4, f'<label for="{ident}">{escape(etiqueta)}</label>')
    else:
        extra = ''
        if 'min' in mods: extra += f' min="{mods["min"]}"'
        if 'max' in mods: extra += f' max="{mods["max"]}"'
        if tipo != 'fecha':
            extra += f' placeholder="{escape(generar_placeholder(tipo, etiqueta))}"'
        emitir(4, f'<label for="{ident}">{escape(etiqueta)}</label>')
        emitir(4, f'<input type="{TIPOS_HTML[tipo]}" id="{ident}" name="{ident}"{extra}{req}>')
    emitir(3, '</div>')

def p_modificadores(p):
    'modificadores : modificadores modificador'
    p[0] = p[1]
    clave, valor = p[2]
    if clave in p[0]:
        error_semantico(f"el modificador '{clave}' está repetido", p.lineno(2) or p.lexer.lineno)
    p[0][clave] = valor

def p_modificadores_vacio(p):
    'modificadores : '
    p[0] = {}

def p_modificador_requerido(p):
    'modificador : REQUERIDO'
    p[0] = ('requerido', True)

def p_modificador_min(p):
    'modificador : MIN NUMERO'
    p[0] = ('min', p[2])

def p_modificador_max(p):
    'modificador : MAX NUMERO'
    p[0] = ('max', p[2])

def p_modificador_lista(p):
    'modificador : CORCHETE_A valores CORCHETE_C'
    p[0] = ('lista', p[2])

def p_valores(p):
    '''valores : valores COMA CADENA
               | CADENA'''
    p[0] = p[1] + [p[3]] if len(p) == 4 else [p[1]]

def p_boton(p):
    'boton : BOTON CADENA'
    emitir(3, f'<button type="submit">{escape(p[2])}</button>')
    emitir(3, '<output></output>')   # acá aparece "Operación realizada"

def p_error(p):
    if p:
        errores.append(f"Error sintáctico: token inesperado '{p.value}' en línea {p.lineno}")
    else:
        errores.append("Error sintáctico: el programa terminó antes de tiempo (¿falta el boton?)")

parser = yacc.yacc(debug=False, write_tables=False)

def traducir(codigo):
    html.clear(); errores.clear(); errores_lexicos.clear(); tabla_campos.clear()
    lexer.lineno = 1
    parser.parse(codigo, lexer=lexer)
    todos = errores_lexicos + errores
    if todos:
        return '\n'.join(todos), False
    pagina = ['<!DOCTYPE html>',
              '<html lang="es">',
              SANGRIA + '<head>',
              SANGRIA * 2 + '<meta charset="utf-8">',
              SANGRIA * 2 + f'<title>{info["titulo"]}</title>',
              SANGRIA + '</head>',
              SANGRIA + '<body>'] + html + [SANGRIA + '</body>', '</html>']
    return '\n'.join(pagina) + '\n', True
````
