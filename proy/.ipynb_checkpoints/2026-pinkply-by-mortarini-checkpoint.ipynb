# pinkpy

## Objetivo

* Diseñar e implementar un lenguaje de programación propio denominado PinkPy, utilizando Python como lenguaje de implementación y Python como lenguaje de destino.
* Desarrollar un traductor capaz de analizar programas escritos en PinkPy mediante un analizador léxico, un analizador sintáctico y un analizador semántico asistido por un agente inteligente. Finalmente, el traductor genera código equivalente en Python.

## Alcance

- Declarar variables de tipo entero, decimal, texto y lógico.
- Asignar valores a variables previamente declaradas.
- Realizar operaciones aritméticas de suma, resta, multiplicación, división y módulo.
- Realizar comparaciones entre expresiones.
- Utilizar operadores lógicos AND, OR y NOT, representados mediante las palabras reservadas y, o y no.
- Mostrar información mediante la instrucción mostrar.
- Construir estructuras condicionales mediante si, sino, entonces y fin.
- Implementar estructuras repetitivas mediante mientras, hacer y fin.
- Utilizar comentarios de una línea.
- Reconocer y detectar errores léxicos, sintácticos y semánticos.
- Traducir las instrucciones válidas de pinkPy a instrucciones equivalentes en Python.


## Especificaciones léxicas

* El análisis léxico de PinkPy se implementa utilizando la biblioteca PLY Lex de Python.
* Su función es leer el código fuente y reconocer los elementos que componen el lenguaje, clasificándolos en tokens.
* Se reconocen palabras reservadas, identificadores, tipos de datos, números, cadenas de texto, operadores y signos de puntuación.

| Token | Lexema | Patrón |
|---|---|---|
| TYPE | entero, decimal, texto, logico | Palabras reservadas |
| SHOW | mostrar | `mostrar` |
| IF | si | `si` |
| ELSE | sino | `sino` |
| THEN | entonces | `entonces` |
| WHILE | mientras | `mientras` |
| DO | hacer | `hacer` |
| END | fin | `fin` |
| AND | y | `y` |
| OR | o | `o` |
| NOT | no | `no` |
| BOOLEAN | verdadero, falso | `verdadero\|falso` |
| IDENTIFIER | edad, nombre, contador | `[a-zA-Z_][a-zA-Z_0-9]*` |
| INTEGER | 20, 150, 0 | `\d+` |
| DECIMAL | 1.75, 3.14 | `\d+\.\d+` |
| STRING | "Hola", "Ana" | `"([^\\\n]|(\\.))*?"` |
| ASSIGN | = | `=` |
| EQUAL | == | `==` |
| NOT_EQUAL | != | `!=` |
| LESS | < | `<` |
| GREATER | > | `>` |
| LESS_EQUAL | <= | `<=` |
| GREATER_EQUAL | >= | `>=` |
| PLUS | + | `\+` |
| MINUS | - | `-` |
| STAR | * | `\*` |
| SLASH | / | `/` |
| PERCENT | % | `%` |
| LPAREN | ( | `\(` |
| RPAREN | ) | `\)` |
| SEMICOLON | ; | `;` |

Los espacios y tabulaciones se ignoran durante el análisis. También se reconocen los comentarios de una línea iniciados con `//`, que son descartados y no generan tokens.

````python
import ply.lex as lex
import json

# Palabras reservadas de PinkPy
reserved = {
    'entero': 'TYPE',
    'decimal': 'TYPE',
    'texto': 'TYPE',
    'logico': 'TYPE',
    'mostrar': 'SHOW',
    'si': 'IF',
    'sino': 'ELSE',
    'entonces': 'THEN',
    'mientras': 'WHILE',
    'hacer': 'DO',
    'fin': 'END',
    'y': 'AND',
    'o': 'OR',
    'no': 'NOT',
    'verdadero': 'BOOLEAN',
    'falso': 'BOOLEAN'
}

# Lista de tokens
tokens = [
    'IDENTIFIER',
    'INTEGER',
    'DECIMAL',
    'STRING',
    'ASSIGN',
    'EQUAL',
    'NOT_EQUAL',
    'LESS',
    'GREATER',
    'LESS_EQUAL',
    'GREATER_EQUAL',
    'PLUS',
    'MINUS',
    'STAR',
    'SLASH',
    'PERCENT',
    'LPAREN',
    'RPAREN',
    'SEMICOLON'
] + list(set(reserved.values()))
t_ignore = ' \t'

t_LESS_EQUAL = r'<='
t_GREATER_EQUAL = r'>='
t_EQUAL = r'=='
t_NOT_EQUAL = r'!='

t_ASSIGN = r'='
t_LESS = r'<'
t_GREATER = r'>'
t_PLUS = r'\+'
t_MINUS = r'-'
t_STAR = r'\*'
t_SLASH = r'/'
t_PERCENT = r'%'
t_LPAREN = r'\('
t_RPAREN = r'\)'
t_SEMICOLON = r';'

# Reconocer números decimales
def t_DECIMAL(t):
    r'\d+\.\d+'
    t.value = float(t.value)
    return t

# Reconocer números enteros
def t_INTEGER(t):
    r'\d+'
    t.value = int(t.value)
    return t

# Reconocer cadenas de texto
def t_STRING(t):
    r'"([^\\\n]|(\\.))*?"'
    t.value = json.loads(t.value)
    return t

# Reconocer identificadores y palabras reservadas
def t_IDENTIFIER(t):
    r'[a-zA-Z_][a-zA-Z_0-9]*'
    t.type = reserved.get(t.value, 'IDENTIFIER')
    return t

# Ignorar comentarios
def t_COMMENT(t):
    r'//.*'
    pass

# Contabilizar saltos de línea
def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    raise SyntaxError(
        f"Carácter no reconocido: {t.value[0]!r}, "
        f"línea {t.lineno}"
    )

# Identificar el archivo del analizador en Jupyter
__file__ = "pinkpy_lexer.py"

# Construir el analizador léxico
lexer = lex.lex(optimize=True)

print("Analizador léxico de PinkPy creado correctamente.")
````

````python
codigo = '''
entero edad = 20;
mostrar(edad);
'''

lexer.input(codigo)

print("TOKENS RECONOCIDOS:\n")

for token in lexer:
    print(token)
````

## Especificaciones sintácticas

* El analizador sintáctico de PinkPy se implementa utilizando la biblioteca PLY Yacc de Python.
* Su función es verificar que la secuencia de tokens obtenida por el analizador léxico respete las reglas gramaticales del lenguaje.

```text
<PROGRAMA> ::= <LISTA_INSTRUCCIONES>

<LISTA_INSTRUCCIONES> ::= <LISTA_INSTRUCCIONES> <INSTRUCCION>
                        | ε

<INSTRUCCION> ::= <DECLARACION>
                | <ASIGNACION>
                | <MOSTRAR>
                | <CONDICIONAL>
                | <CICLO>

<DECLARACION> ::= TYPE IDENTIFIER ASSIGN <EXPRESION> SEMICOLON

<ASIGNACION> ::= IDENTIFIER ASSIGN <EXPRESION> SEMICOLON

<MOSTRAR> ::= SHOW LPAREN <EXPRESION> RPAREN SEMICOLON

<TIPO> ::= TYPE

<CONDICIONAL> ::= IF <EXPRESION> THEN
                  <LISTA_INSTRUCCIONES>
                  <ALTERNATIVA>
                  END

<ALTERNATIVA> ::= ELSE <LISTA_INSTRUCCIONES>
                | ε

<CICLO> ::= WHILE <EXPRESION> DO
            <LISTA_INSTRUCCIONES>
            END

<EXPRESION> ::= <EXPRESION_OR>

<EXPRESION_OR> ::= <EXPRESION_OR> OR <EXPRESION_AND>
                 | <EXPRESION_AND>

<EXPRESION_AND> ::= <EXPRESION_AND> AND <EXPRESION_IGUALDAD>
                  | <EXPRESION_IGUALDAD>

<EXPRESION_IGUALDAD> ::= <EXPRESION_IGUALDAD> EQUAL <EXPRESION_COMPARACION>
                        | <EXPRESION_IGUALDAD> NOT_EQUAL <EXPRESION_COMPARACION>
                        | <EXPRESION_COMPARACION>

<EXPRESION_COMPARACION> ::= <EXPRESION_COMPARACION> LESS <EXPRESION_TERMINO>
                           | <EXPRESION_COMPARACION> GREATER <EXPRESION_TERMINO>
                           | <EXPRESION_COMPARACION> LESS_EQUAL <EXPRESION_TERMINO>
                           | <EXPRESION_COMPARACION> GREATER_EQUAL <EXPRESION_TERMINO>
                           | <EXPRESION_TERMINO>

<EXPRESION_TERMINO> ::= <EXPRESION_TERMINO> PLUS <EXPRESION_FACTOR>
                       | <EXPRESION_TERMINO> MINUS <EXPRESION_FACTOR>
                       | <EXPRESION_FACTOR>

<EXPRESION_FACTOR> ::= <EXPRESION_FACTOR> STAR <EXPRESION_UNARIA>
                      | <EXPRESION_FACTOR> SLASH <EXPRESION_UNARIA>
                      | <EXPRESION_FACTOR> PERCENT <EXPRESION_UNARIA>
                      | <EXPRESION_UNARIA>

<EXPRESION_UNARIA> ::= MINUS <EXPRESION_UNARIA>
                      | NOT <EXPRESION_UNARIA>
                      | <EXPRESION_PRIMARIA>

<EXPRESION_PRIMARIA> ::= INTEGER
                        | DECIMAL
                        | STRING
                        | BOOLEAN
                        | IDENTIFIER
                        | LPAREN <EXPRESION> RPAREN
```

````python

import ply.yacc as yacc
start = 'programa'


def p_programa(p):
    'programa : lista_instrucciones'
    p[0] = p[1]


def p_lista_instrucciones(p):
    '''lista_instrucciones : lista_instrucciones instruccion
                           | empty'''
    if len(p) == 3:
        p[0] = p[1] + [p[2]]
    else:
        p[0] = []


def p_instruccion(p):
    '''instruccion : declaracion
                   | asignacion
                   | mostrar
                   | condicional
                   | ciclo'''
    p[0] = p[1]

# declaraciones

def p_declaracion(p):
    'declaracion : TYPE IDENTIFIER ASSIGN expresion SEMICOLON'
    p[0] = {
        'tipo': 'declaracion',
        'tipo_dato': p[1],
        'variable': p[2],
        'valor': p[4]
    }


def p_asignacion(p):
    'asignacion : IDENTIFIER ASSIGN expresion SEMICOLON'
    p[0] = {
        'tipo': 'asignacion',
        'variable': p[1],
        'valor': p[3]
    }


def p_mostrar(p):
    'mostrar : SHOW LPAREN expresion RPAREN SEMICOLON'
    p[0] = {
        'tipo': 'mostrar',
        'expresion': p[3]
    }


# ESTRUCTURAS DE CONTROL

def p_condicional(p):
    'condicional : IF expresion THEN lista_instrucciones alternativa END'
    p[0] = {
        'tipo': 'condicional',
        'condicion': p[2],
        'entonces': p[4],
        'sino': p[5]
    }



def p_alternativa(p):
    '''alternativa : ELSE lista_instrucciones
                   | empty'''
    if len(p) == 3:
        p[0] = p[2]
    else:
        p[0] = []


def p_ciclo(p):
    'ciclo : WHILE expresion DO lista_instrucciones END'
    p[0] = {
        'tipo': 'ciclo',
        'condicion': p[2],
        'instrucciones': p[4]
    }


# EXPRESIONES Y PRECEDENCIA

def p_expresion(p):
    'expresion : expresion_or'
    p[0] = p[1]


def p_expresion_or_binaria(p):
    '''expresion_or : expresion_or OR expresion_and
                    | expresion_and'''
    if len(p) == 4:
        p[0] = {
            'tipo': 'binaria',
            'operador': p[2],
            'izquierda': p[1],
            'derecha': p[3]
        }
    else:
        p[0] = p[1]


def p_expresion_and_binaria(p):
    '''expresion_and : expresion_and AND expresion_igualdad
                     | expresion_igualdad'''
    if len(p) == 4:
        p[0] = {
            'tipo': 'binaria',
            'operador': p[2],
            'izquierda': p[1],
            'derecha': p[3]
        }
    else:
        p[0] = p[1]


def p_expresion_igualdad(p):
    '''expresion_igualdad : expresion_igualdad EQUAL expresion_comparacion
                          | expresion_igualdad NOT_EQUAL expresion_comparacion
                          | expresion_comparacion'''
    if len(p) == 4:
        p[0] = {
            'tipo': 'binaria',
            'operador': p[2],
            'izquierda': p[1],
            'derecha': p[3]
        }
    else:
        p[0] = p[1]


def p_expresion_comparacion(p):
    '''expresion_comparacion : expresion_comparacion LESS expresion_termino
                             | expresion_comparacion GREATER expresion_termino
                             | expresion_comparacion LESS_EQUAL expresion_termino
                             | expresion_comparacion GREATER_EQUAL expresion_termino
                             | expresion_termino'''
    if len(p) == 4:
        p[0] = {
            'tipo': 'binaria',
            'operador': p[2],
            'izquierda': p[1],
            'derecha': p[3]
        }
    else:
        p[0] = p[1]


def p_expresion_termino(p):
    '''expresion_termino : expresion_termino PLUS expresion_factor
                         | expresion_termino MINUS expresion_factor
                         | expresion_factor'''
    if len(p) == 4:
        p[0] = {
            'tipo': 'binaria',
            'operador': p[2],
            'izquierda': p[1],
            'derecha': p[3]
        }
    else:
        p[0] = p[1]


def p_expresion_factor(p):
    '''expresion_factor : expresion_factor STAR expresion_unaria
                        | expresion_factor SLASH expresion_unaria
                        | expresion_factor PERCENT expresion_unaria
                        | expresion_unaria'''
    if len(p) == 4:
        p[0] = {
            'tipo': 'binaria',
            'operador': p[2],
            'izquierda': p[1],
            'derecha': p[3]
        }
    else:
        p[0] = p[1]


def p_expresion_unaria(p):
    '''expresion_unaria : MINUS expresion_unaria
                        | NOT expresion_unaria'''
    p[0] = {
        'tipo': 'unaria',
        'operador': p[1],
        'expresion': p[2]
    }


def p_expresion_unaria_primaria(p):
    'expresion_unaria : expresion_primaria'
    p[0] = p[1]


def p_expresion_primaria_literal(p):
    '''expresion_primaria : INTEGER
                          | DECIMAL
                          | STRING
                          | BOOLEAN'''
    p[0] = {
        'tipo': 'literal',
        'valor': p[1]
    }


def p_expresion_primaria_variable(p):
    'expresion_primaria : IDENTIFIER'
    p[0] = {
        'tipo': 'variable',
        'nombre': p[1]
    }


def p_expresion_primaria_parentesis(p):
    'expresion_primaria : LPAREN expresion RPAREN'
    p[0] = p[2]


# PRODUCCIÓN VACÍA


def p_empty(p):
    'empty :'
    pass

# MANEJO DE ERRORES


def p_error(p):
    if p:
        raise SyntaxError(
            f"Error sintáctico cerca de {p.value!r}, "
            f"línea {p.lineno}"
        )
    else:
        
        raise SyntaxError(
            "Error sintáctico: fin inesperado del programa"
        )

````

## Especificaciones semánticas

* Se utilizo el agente Qwen ejecutado localmente mediante Ollama.
* El agente recibe el codigo fuente de pinkpy y el arbol AST y luego comprueba que se cumplan las reglas 

| Regla gramatical | Acción semántica |
|---|---|
| PROGRAMA | Analizar el conjunto de instrucciones y verificar que el programa respete las reglas semánticas. |
| LISTA_INSTRUCCIONES | Analizar las instrucciones en el orden en que aparecen y considerar las declaraciones anteriores. |
| INSTRUCCION | Identificar el tipo de instrucción y aplicar las validaciones correspondientes. |
| DECLARACION | Verificar que la variable no haya sido declarada previamente y que el valor sea compatible con el tipo indicado. |
| ASIGNACION | Comprobar que la variable haya sido declarada y que el valor asignado sea compatible con su tipo. |
| MOSTRAR | Verificar que la expresión sea válida y que las variables utilizadas estén declaradas. |
| CONDICIONAL | Comprobar que la condición sea de tipo lógico y validar las instrucciones de sus bloques. |
| ALTERNATIVA | Validar las instrucciones correspondientes al bloque sino. |
| CICLO | Comprobar que la condición sea de tipo lógico y validar las instrucciones del ciclo. |
| EXPRESION_OR | Verificar que los operandos de la operación o sean lógicos. |
| EXPRESION_AND | Verificar que los operandos de la operación y sean lógicos. |
| EXPRESION_IGUALDAD | Comprobar que los operandos de igualdad o desigualdad sean compatibles. |
| EXPRESION_COMPARACION | Verificar la compatibilidad de los operandos de comparación. |
| EXPRESION_TERMINO | Comprobar que los operandos de suma y resta sean numéricos. |
| EXPRESION_FACTOR | Comprobar que los operandos de multiplicación, división y módulo sean numéricos. |
| EXPRESION_UNARIA | Verificar que la negación aritmética se aplique a números y que no se aplique a valores lógicos. |
| EXPRESION_PRIMARIA | Reconocer valores constantes y variables, verificando que las variables hayan sido declaradas previamente. |




## Implementación

* La implementación del traductor PinkPy se desarrolló utilizando Python y las bibliotecas PLY Lex y PLY Yacc para realizar el análisis léxico y sintáctico.
* Analizador léxico:
    * Se implementó mediante PLY Lex. Su función es leer el código fuente escrito en PinkPy, reconocer los elementos del lenguaje y clasificarlos en tokens, como palabras reservadas, identificadores, números, cadenas de texto y operadores.
* Analizador sintáctico:
    * Se desarrolló utilizando PLY Yacc. Recibe los tokens generados por el analizador léxico, comprueba que las instrucciones respeten la gramática BNF definida para PinkPy y construye un árbol sintáctico (AST).
* Analizador semántico:
    * Se implementó mediante un agente inteligente conectado a Ollama, que ejecuta localmente el modelo Qwen 2.5:3b. Recibe el código fuente y el árbol sintáctico para verificar la declaración de variables, la compatibilidad de los tipos de datos y la coherencia de las operaciones.
* Generador de código:
    * Se desarrolló en Python. Su función es recorrer el árbol sintáctico y transformar las instrucciones válidas de PinkPy en código equivalente en Python. La generación se realiza mediante reglas definidas en el propio programa.
* Ejecución del traductor:
    * Una vez completadas las etapas de análisis, si el programa supera las validaciones correspondientes, se genera el código Python resultante. Este código puede ejecutarse para comprobar el funcionamiento de la traducción.

### AGENTE

````python

import json
import requests

def analizar_semantica(codigo, arbol):

    print("Agente semántico de PinkPy")
    instrucciones = f"""

    REGLAS SEMÁNTICAS OBLIGATORIAS:

    1. Los tipos disponibles son entero, decimal, texto y logico.
    2. Las variables deben declararse antes de utilizarse.
    3. Una variable debe recibir valores compatibles con su tipo.
    4. Una variable entero acepta valores enteros.
    5. Una variable decimal acepta valores decimales y enteros.
    6. Una variable texto acepta valores de tipo texto.
    7. Una variable logico acepta verdadero o falso.
    8. Las operaciones matemáticas requieren valores numéricos.
    9. Las condiciones de si y mientras deben ser lógicas.
    10. Los operadores y, o y no requieren valores lógicos.
    11. La instrucción mostrar() acepta expresiones de cualquier tipo.
    12. Mostrar una variable previamente declarada es válido.
    

IMPORTANTE SOBRE LAS DECLARACIONES:

Analizá las instrucciones en el orden en que aparecen.

Cuando encuentres una declaración válida, registrá la variable
como declarada desde ese punto en adelante.

Una variable declarada puede utilizarse en las instrucciones
siguientes del mismo programa.

Ejemplo válido:
texto nombre = "Ana";
mostrar(nombre);

En este ejemplo, nombre SÍ está declarada antes de utilizarse.
El resultado debe ser:
{{"valido": true, "errores": []}}

No informes variables como no declaradas si aparecen
previamente en una declaración válida.

    IMPORTANTE:
    No inventes restricciones que no estén en estas reglas.
    No consideres que mostrar() exige texto o logico.
Ejemplo de declaración válida:
decimal precio = 15.50;
Resultado:
{{"valido": true, "errores": []}}
    Analizá el código PinkPy y su árbol sintáctico.

    Respondé exclusivamente en JSON:
    {{
        "valido": true,
        "errores": []
    }}

    Si hay errores, valido debe ser false y errores debe
    describirlos.

    Código PinkPy:
    {codigo}

    Árbol sintáctico:
    {json.dumps(arbol, ensure_ascii=False)}
    """
    respuesta = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": "qwen2.5:3b",
            "messages": [
                {
                    "role": "user",
                    "content": instrucciones
                }
            ],
            "format": "json",
            "stream": False
        },
        timeout=180
    )

    respuesta.raise_for_status()

    resultado = json.loads(
        respuesta.json()["message"]["content"]
    )

    return resultado



````

````python

codigo_prueba = """
entero edad = 20;
mostrar(edad);
"""


lexer.input(codigo_prueba)
arbol = parser.parse(codigo_prueba, lexer=lexer)

resultado = analizar_semantica(codigo_prueba, arbol)

print("RESULTADO DEL ANÁLISIS SEMÁNTICO:")
print(json.dumps(resultado, indent=4, ensure_ascii=False))
````

### GENERADOR

````python

def generar_expresion(nodo):

    if nodo["tipo"] == "literal":
        valor = nodo["valor"]

        if valor == "verdadero":
            return "True"
        elif valor == "falso":
            return "False"

        return repr(valor)

    elif nodo["tipo"] == "variable":
        return nodo["nombre"]

    elif nodo["tipo"] == "binaria":
        izquierda = generar_expresion(nodo["izquierda"])
        derecha = generar_expresion(nodo["derecha"])

        operadores = {
            "y": "and",
            "o": "or"
        }

        operador = operadores.get(
            nodo["operador"],
            nodo["operador"]
        )

        return f"({izquierda} {operador} {derecha})"

    elif nodo["tipo"] == "unaria":
        expresion = generar_expresion(nodo["expresion"])

        operador = {
            "no": "not",
            "-": "-"
        }.get(nodo["operador"], nodo["operador"])

        return f"({operador} {expresion})"

    else:
        raise ValueError(
            f"Expresión desconocida: {nodo}"
        )


def generar_instrucciones(instrucciones, nivel=0):

    codigo = []
    sangria = "    " * nivel

    for instruccion in instrucciones:

        tipo = instruccion["tipo"]

        if tipo == "declaracion":
            variable = instruccion["variable"]
            valor = generar_expresion(instruccion["valor"])
            codigo.append(f"{sangria}{variable} = {valor}")

        elif tipo == "asignacion":
            variable = instruccion["variable"]
            valor = generar_expresion(instruccion["valor"])
            codigo.append(f"{sangria}{variable} = {valor}")

        elif tipo == "mostrar":
            expresion = generar_expresion(instruccion["expresion"])
            codigo.append(f"{sangria}print({expresion})")

        elif tipo == "condicional":
            condicion = generar_expresion(instruccion["condicion"])
            codigo.append(f"{sangria}if {condicion}:")

            entonces = instruccion["entonces"]

            if entonces:
                codigo.extend(
                    generar_instrucciones(entonces, nivel + 1)
                )
            else:
                codigo.append(f"{sangria}    pass")

            alternativa = instruccion["sino"]

            if alternativa:
                codigo.append(f"{sangria}else:")
                codigo.extend(
                    generar_instrucciones(alternativa, nivel + 1)
                )

        elif tipo == "ciclo":
            condicion = generar_expresion(instruccion["condicion"])
            codigo.append(f"{sangria}while {condicion}:")

            cuerpo = instruccion["instrucciones"]

            if cuerpo:
                codigo.extend(
                    generar_instrucciones(cuerpo, nivel + 1)
                )
            else:
                codigo.append(f"{sangria}    pass")

    return codigo


def generar_codigo_python(arbol):

    instrucciones = generar_instrucciones(arbol)

    if not instrucciones:
        return "pass"

    return "\n".join(instrucciones)

````

````python

codigo_prueba = """
entero edad = 20;
mostrar(edad);
"""

lexer.input(codigo_prueba)
arbol = parser.parse(codigo_prueba, lexer=lexer)

resultado = analizar_semantica(codigo_prueba, arbol)

if resultado["valido"]:
    codigo_python = generar_codigo_python(arbol)

    print("TRADUCCIÓN A PYTHON:")
    print(codigo_python)
else:
    print("ERROR SEMÁNTICO:")
    print(resultado["errores"])
````
