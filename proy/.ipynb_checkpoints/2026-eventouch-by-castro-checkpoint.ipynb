# eventouch

## Objetivo

 Simplificar mediante eventouch, desde lenguaje natural en español, los eventos de puntero y ratón a código JavaScript.

## Alcance

* **Soporte para eventos de interacción web**:
  * **7 eventos compartidos/equivalentes** entre Pointer y Mouse: presionar (`pointerdown`/`mousedown`), soltar (`pointerup`/`mouseup`), moverse (`pointermove`/`mousemove`), estoy (`pointerover`/`mouseover`), no estoy (`pointerout`/`mouseout`), ingreso (`pointerenter`/`mouseenter`) y salgo (`pointerleave`/`mouseleave`).
  * **4 eventos exclusivos de Pointer Events**: cancelar (`pointercancel`), capturar puntero (`gotpointercapture`), perder captura (`lostpointercapture`) y actualización de puntero (`pointerrawupdate`).
  * **5 eventos exclusivos de Mouse / Interfaz**: clic (`click`), doble clic (`dblclick`), clic auxiliar (`auxclick`), rueda (`wheel`) y menú contextual (`contextmenu`).
* **Diferenciación de dispositivo**: Soporte opcional para indicar explícitamente `con puntero` o `con mouse` en los eventos compartidos, usando el modelo unificado de Pointer Events por defecto.
* **Bloques de Acciones de Manipulación del DOM**:
  * Delimitados mediante el token `DP` (`:` dos puntos) luego del selector disparador, permitiendo encadenar múltiples acciones sobre distintos elementos identificados por `VALOR`:
    * `CAMBIAR`: Modifica propiedades visuales o de contenido (`TEXTO`, `COLOR`, `FONDO`, `TAMAÑO`, `OPACIDAD`) de un elemento.
    * `MOSTRAR` / `OCULTAR`: Controla la visibilidad de un elemento.
    * `HABILITAR` / `DESHABILITAR`: Controla el estado interactivo de un elemento.
* **Análisis Léxico y Sintáctico**:
  * Reconocimiento de palabras clave en lenguaje natural en español, selectores CSS (`#id`, `.clase`), delimitador de bloque `DP` (`:`) y literales entre comillas (`VALOR`).
  * Detección y reporte de errores léxicos (caracteres no reconocidos) y errores sintácticos (estructura de oración inválida).
* **Traducción Semántica con LLM**:
  * Generación automática de código JavaScript estándar utilizando `document.querySelector()`, `document.getElementById()` y `addEventListener()`.
* **Limitaciones**:
  * No ejecuta el código JavaScript resultante en un navegador real, sino que genera el archivo `.js` listo para ser integrado en un proyecto web.

| :--- | :--- | :--- | :--- |

| `con puntero`, `con pointer` | `MODO_POINTER` | `[Cc]on[ \t]+(puntero\|pointer)` | Modificador de dispositivo |
| `con mouse`, `con raton` | `MODO_MOUSE` | `[Cc]on[ \t]+(mouse\|rat[óo]n)` | Modificador de dispositivo |
| `presiono` | `DOWN` | `[Pp]resiono` | Evento compartido  |
| `suelto` | `UP` | `[Ss]uelto` | Evento compartido  |
| `me muevo` | `MOVE` | `[Mm]e[ \t]+muevo` | Evento compartido  |
| `ingreso` | `ENTER` | `[Ii]ngreso` | Evento compartido  |
| `salgo` | `LEAVE` | `[Ss]algo` | Evento compartido  |
| `estoy` | `OVER` | `[Ee]stoy` | Evento compartido  |
| `no estoy` | `OUT` | `[Nn]o[ \t]+estoy` | Evento compartido  |
| `cancelo puntero` | `CANCEL` | `[Cc]ancelo[ \t]+puntero` | Exclusivo Pointer  |
| `capturo puntero` | `GOTCAPTURE` | `[Cc]apturo[ \t]+puntero` | Exclusivo Pointer  |
| `pierdo captura` | `LOSTCAPTURE` | `[Pp]ierdo[ \t]+captura` | Exclusivo Pointer  |
| `detecto puntero` | `POINTER_RAW` | `[Dd]etecto[ \t]+puntero` | Exclusivo Pointer  |
| `doble clic` | `DBLCLICK` | `[Dd]oble[ \t]+clic` | Exclusivo Mouse  |
| `clic auxiliar` | `AUXCLICK` | `[Cc]lic[ \t]+auxiliar` | Exclusivo Mouse  |
| `hago clic`, `clic` | `CLICK` | `[Hh]ago[ \t]+clic\|[Cc]lic` | Exclusivo Mouse  |
| `giro rueda`, `rueda` | `WHEEL` | `[Gg]iro[ \t]+rueda\|[Rr]ueda` | Exclusivo Mouse  |
| `menu contextual` | `CONTEXTMENU` | `[Mm]en[úu][ \t]+contextual` | Exclusivo Mouse  |
| `sobre` | `SOBRE` | `[Ss]obre` | Preposición de enlace |
| `#boton`, `.tarjeta` | `SELECTOR` | `[\#\.][a-zA-Z0-9_\-]+` | Selector CSS disparador |
| `:` | `DP` | `:` | Delimitador de inicio de bloque de acciones |
| `CAMBIAR` | `CAMBIAR` | `[Cc]ambiar` | Acción de modificación |
| `MOSTRAR` | `MOSTRAR` | `[Mm]ostrar` | Acción de visibilidad |
| `OCULTAR` | `OCULTAR` | `[Oo]cultar` | Acción de visibilidad |
| `HABILITAR` | `HABILITAR` | `[Hh]abilitar` | Acción de estado |
| `DESHABILITAR` | `DESHABILITAR` | `[Dd]eshabilitar` | Acción de estado |
| `TEXTO`, `COLOR`, `FONDO`, `TAMAÑO`, `OPACIDAD` | `PROPIEDAD` | `[Tt]exto\|[Cc]olor\|[Ff]ondo\|[Tt]ama[ñn]o\|[Oo]pacidad` | Propiedad modificable |
| `"Correcto"`, `"verde"`, `"resultado"` | `VALOR` | `\"[^\"]+\"` | Valor asignado o id del elemento |

## Especificaciones sintácticas

*(El token `DP` corresponde al símbolo `:` [dos puntos], marcando el inicio del bloque de `<Acciones>`. Se permite producción vacía $\lambda$ para bloques de acciones opcionales).* 

```bnf
<Programa> ::= <Instruccion> | <Instruccion> <Programa>

<Instruccion> ::= CUANDO <Evento> SOBRE SELECTOR DP <Acciones>
<Instruccion> ::= CUANDO <Evento> <Dispositivo> SOBRE SELECTOR DP <Acciones>

<Dispositivo> ::= MODO_POINTER
<Dispositivo> ::= MODO_MOUSE

<Acciones> ::= <Accion> <Acciones> | λ

<Accion> ::= CAMBIAR <Elemento> PROPIEDAD VALOR
<Accion> ::= MOSTRAR <Elemento>
<Accion> ::= OCULTAR <Elemento>
<Accion> ::= HABILITAR <Elemento>
<Accion> ::= DESHABILITAR <Elemento>

<Elemento> ::= VALOR

<Evento> ::= DOWN | UP | MOVE | OVER | OUT | ENTER | LEAVE
           | CANCEL | GOTCAPTURE | LOSTCAPTURE | POINTER_RAW
           | CLICK | DBLCLICK | AUXCLICK | WHEEL | CONTEXTMENU
```

## Especificaciones semánticas

| Regla GIC | Acciones Semánticas |
| :--- | :--- |
| `<Programa> ::= <Instruccion>` | Procesa una única instrucción completa (evento + delimitador `DP` + bloque de acciones) y la almacena en la lista de salida JS. |
| `<Programa> ::= <Instruccion> <Programa>` | Procesa múltiples instrucciones secuenciales acumulando cada `addEventListener` generado. |
| `<Instruccion> ::= CUANDO <Evento> SOBRE SELECTOR DP <Acciones>` | Asume modo unificado moderno (`pointer*` para eventos compartidos o el evento específico de mouse o pointer), reconoce `DP` (`:`) como apertura del bloque `<Acciones>` e invoca al agente LLM para construir el `addEventListener`. |
| `<Instruccion> ::= CUANDO <Evento> <Dispositivo> SOBRE SELECTOR DP <Acciones>` | Si `<Dispositivo>` es `MODO_MOUSE`, fuerza la traducción al equivalente clásico (`mousedown`, `mousemove`, etc.); si es `MODO_POINTER`, fuerza `pointer*`. Pasa las `<Acciones>` al LLM. |
| `<Dispositivo> ::= MODO_POINTER \| MODO_MOUSE` | Retorna el modificador de dispositivo detectado para condicionar la traducción. |
| `<Acciones> ::=  <Accion> <Acciones> \| λ` | Construye recursivamente la lista ordenada de diccionarios de acciones a ejecutar dentro del callback del evento. |
| `<Accion> ::= CAMBIAR <Elemento> PROPIEDAD VALOR` | Registra una operación de modificación de estilo o contenido (`TEXTO`, `COLOR`, `FONDO`, `TAMAÑO`, `OPACIDAD`) sobre el elemento identificado, asignándole un `VALOR`. |
| `<Accion> ::= MOSTRAR <Elemento> \| OCULTAR <Elemento> \| HABILITAR <Elemento> \| DESHABILITAR <Elemento>` | Registra una operación unaria de visibilidad (`display`) o estado interactivo (`disabled`) sobre el elemento. |
| `<Elemento> ::= VALOR` | Retorna el id del elemento del DOM sobre el que actúa la acción. |
| `<Evento> ::= ...` | Retorna el token reconocido, que se usa para determinar el evento DOM. |

## Implementación

````python
%%file traductor/lexer_eventouch.py
import ply.lex as lex

tokens = (
    'CUANDO',
    'MODO_POINTER',
    'MODO_MOUSE',
    'DOWN',
    'UP',
    'MOVE',
    'ENTER',
    'LEAVE',
    'OVER',
    'OUT',
    'CANCEL',
    'GOTCAPTURE',
    'LOSTCAPTURE',
    'POINTER_RAW',
    'DBLCLICK',
    'AUXCLICK',
    'CLICK',
    'WHEEL',
    'CONTEXTMENU',
    'SOBRE',
    'SELECTOR',
    'DP',
    'CAMBIAR',
    'MOSTRAR',
    'OCULTAR',
    'HABILITAR',
    'DESHABILITAR',
    'PROPIEDAD',
    'VALOR'
)

errores_lexicos = []

def t_CUANDO(t):
    r'[Cc]uando'
    return t

def t_MODO_POINTER(t):
    r'[Cc]on[ \t]+(puntero|pointer)'
    return t

def t_MODO_MOUSE(t):
    r'[Cc]on[ \t]+(mouse|rat[óo]n)'
    return t

# Eventos exclusivos de Pointer (4)
def t_CANCEL(t):
    r'[Cc]ancelo[ \t]+puntero'
    return t

def t_GOTCAPTURE(t):
    r'[Cc]apturo[ \t]+puntero'
    return t

def t_LOSTCAPTURE(t):
    r'[Pp]ierdo[ \t]+captura'
    return t

def t_POINTER_RAW(t):
    r'[Dd]etecto[ \t]+puntero'
    return t

# Eventos exclusivos de Mouse (5)
def t_DBLCLICK(t):
    r'[Dd]oble[ \t]+clic'
    return t

def t_AUXCLICK(t):
    r'[Cc]lic[ \t]+auxiliar'
    return t

def t_CLICK(t):
    r'[Hh]ago[ \t]+clic|[Cc]lic'
    return t

def t_WHEEL(t):
    r'[Gg]iro[ \t]+rueda|[Rr]ueda'
    return t

def t_CONTEXTMENU(t):
    r'[Mm]en[úu][ \t]+contextual'
    return t

# Eventos compartidos entre Pointer y Mouse (7)

def t_ENTER(t):
    r'\b[Ii]ngreso\b'
    return t

def t_LEAVE(t):
    r'\b[Ss]algo\b'
    return t

def t_OVER(t):
    r'\b[Ee]stoy\b'
    return t

def t_OUT(t):
    r'\b[Nn]o[ \t]+estoy\b'
    return t

def t_DOWN(t):
    r'[Pp]resiono'
    return t

def t_UP(t):
    r'[Ss]uelto'
    return t

def t_MOVE(t):
    r'[Mm]e[ \t]+muevo'
    return t

def t_SOBRE(t):
    r'[Ss]obre'
    return t

def t_SELECTOR(t):
    r'[\#\.][a-zA-Z0-9_\-]+'
    return t

# Delimitador DP (dos puntos) que marca el inicio del bloque de acciones
def t_DP(t):
    r':'
    return t

# Acciones de manipulación de UI (DOM)
def t_CAMBIAR(t):
    r'[Cc]ambiar'
    return t

def t_MOSTRAR(t):
    r'[Mm]ostrar'
    return t

def t_OCULTAR(t):
    r'[Oo]cultar'
    return t

def t_HABILITAR(t):
    r'[Hh]abilitar'
    return t

def t_DESHABILITAR(t):
    r'[Dd]eshabilitar'
    return t

def t_PROPIEDAD(t):
    r'[Tt]exto|[Cc]olor|[Ff]ondo|[Tt]ama[ñn]o|[Oo]pacidad'
    return t

# Nuevo valor del elemento
def t_VALOR(t):
    r'\"[^\"]+\"'
    t.value = t.value.strip('"')
    return t

t_ignore = ' \t'

def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    msg = f"Error léxico: carácter ilegal '{t.value[0]}' en línea {t.lexer.lineno}"
    print(msg)
    errores_lexicos.append(msg)
    t.lexer.skip(1)

lexer = lex.lex()
````

````python
%%file traductor/ag_generar_js.py
import os
from openai import OpenAI

client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"], 
)

def generar_codigo_js(accion_espanol, selector, acciones, dispositivo="unificado"):

    prompt = f'''
    Eres un compilador estricto y optimizado de español a JavaScript.
    Devuelve ÚNICAMENTE el código JavaScript puro.
    Sin explicaciones, sin bloques de código markdown (no uses ```), sin texto adicional.

    Traduce la siguiente regla a JavaScript usando document.querySelector, document.getElementById y addEventListener:
    - Evento disparador en español: "{accion_espanol}"
    - Modo de dispositivo: "{dispositivo}"
    - Selector CSS disparador: "{selector}"
    - Lista de acciones a ejecutar dentro del callback: {acciones}

    REGLAS DE OPTIMIZACIÓN DE CÓDIGO (IMPORTATE) :
    1. Si múltiples acciones dentro del bloque afectan al mismo elemento objetivo (por ejemplo, "resultado"), NO llames repetidamente a document.getElementById() para cada línea.
    2. Guarda el elemento en una constante al inicio del callback si se usa más de una vez (ej: const resultado = document.getElementById("resultado");) y reutiliza esa variable.
    3. Para acciones unísecas o elementos distintos, usa document.getElementById("<elemento>") directamente.

    Reglas de mapeo de eventos exactas:

    1. Si el modo es "con mouse" o "con raton", usa eventos clásicos de ratón (mousedown, mouseup, mousemove, mouseover, mouseout, mouseenter, mouseleave).
    2. Si el modo es "con puntero", "con pointer" o "unificado" para acciones compartidas solo entre pointer y mouse, usa Pointer Events (pointerdown, pointerup, pointermove, pointerover, pointerout, pointerenter, pointerleave).
    3. Exclusivos de Pointer: "cancelo puntero" -> pointercancel, "capturo puntero" -> gotpointercapture, "pierdo captura" -> lostpointercapture, "detecto puntero" -> pointerrawupdate.
    4. Exclusivos de Mouse: "clic" / "hago clic" -> click, "doble clic" -> dblclick, "clic auxiliar" -> auxclick, "rueda" / "giro rueda" -> wheel, "menu contextual" -> contextmenu.
    5. Nunca usar eventos de Keyboard.

    Reglas de mapeo de acciones (cada acción indica su 'elemento' objetivo):
    - CAMBIAR con propiedad TEXTO -> document.getElementById("<elemento>").textContent = "<valor>";
    - CAMBIAR con propiedad COLOR -> document.getElementById("<elemento>").style.color = "<valor>"; (traduce colores en español a CSS válido, ej. "verde" -> "green", "rojo" -> "red", "azul" -> "blue")
    - CAMBIAR con propiedad FONDO -> document.getElementById("<elemento>").style.backgroundColor = "<valor>";
    - CAMBIAR con propiedad TAMAÑO -> document.getElementById("<elemento>").style.fontSize = "<valor>";
    - CAMBIAR con propiedad OPACIDAD -> document.getElementById("<elemento>").style.opacity = "<valor>";
    - MOSTRAR -> document.getElementById("<elemento>").style.display = "block";
    - OCULTAR -> document.getElementById("<elemento>").style.display = "none";
    - HABILITAR -> document.getElementById("<elemento>").disabled = false;
    - DESHABILITAR -> document.getElementById("<elemento>").disabled = true;

    Formato exacto requerido:
    document.querySelector("{selector}").addEventListener("<evento>", (event) => {{
        // sentencias JS optimizado de cada acción en orden
    }});
    '''


    respuesta = client.chat.completions.create(
        model="openai/gpt-oss-20b", 
        messages=[{"role": "user", "content": prompt}],
        temperature=0,
    )
    return respuesta.choices[0].message.content.strip()
````

````python
%%file traductor/parser_eventouch.py
import ply.yacc as yacc
from lexer_eventouch import tokens
from ag_generar_js import generar_codigo_js

js_codigo = []

def p_programa(p):
    '''
    programa : programa instruccion
             | instruccion
    '''
    pass

def p_instruccion_simple(p):
    'instruccion : CUANDO evento SOBRE SELECTOR DP acciones'
    codigo = generar_codigo_js(
        accion_espanol=p[2],
        selector=p[4],
        acciones=p[6],
        dispositivo="unificado"
    )
    js_codigo.append(codigo)

def p_instruccion_con_dispositivo(p):
    'instruccion : CUANDO evento dispositivo SOBRE SELECTOR DP acciones'
    codigo = generar_codigo_js(
        accion_espanol=p[2],
        selector=p[5],
        acciones=p[7],
        dispositivo=p[3].lower()
    )
    js_codigo.append(codigo)

def p_dispositivo(p):
    '''
    dispositivo : MODO_POINTER
                | MODO_MOUSE
    '''
    p[0] = p[1]

def p_acciones(p):
    '''
    acciones : accion acciones
             | lambda_vacio
    '''
    if len(p) == 3:
        p[0] = [p[1]] + p[2]
    else:
        p[0] = []

def p_lambda_vacio(p):
    'lambda_vacio :'
    p[0] = None

def p_accion_cambiar(p):
    'accion : CAMBIAR elemento PROPIEDAD VALOR'
    p[0] = {"tipo": "CAMBIAR", "elemento": p[2], "propiedad": p[3].upper(), "valor": p[4]}

def p_elemento(p):
    'elemento : VALOR'
    p[0] = p[1]


def p_accion_unaria(p):
    '''
    accion : MOSTRAR elemento
           | OCULTAR elemento
           | HABILITAR elemento
           | DESHABILITAR elemento
    '''
    p[0] = {"tipo": p[1].upper(), "elemento": p[2]}

def p_evento(p):
    '''
    evento : DOWN
           | UP
           | MOVE
           | OVER
           | OUT
           | ENTER
           | LEAVE
           | CANCEL
           | GOTCAPTURE
           | LOSTCAPTURE
           | POINTER_RAW
           | CLICK
           | DBLCLICK
           | AUXCLICK
           | WHEEL
           | CONTEXTMENU
    '''
    # p[0] = p[1]

    p[0] = p.slice[1].type 

def p_error(p):
    if p:
        js_codigo.append(f"Error sintáctico en token '{p.value}' (línea {p.lineno})")
    else:
        js_codigo.append("Error sintáctico: fin de archivo inesperado")

parser = yacc.yacc()
````

````python
%%file traductor/main.py
import sys
from lexer_eventouch import lexer, errores_lexicos
from parser_eventouch import parser, js_codigo

def main():
    if len(sys.argv) < 2:
        print("Uso: python main.py <archivo_entrada>")
        sys.exit(1)

    ruta_entrada = sys.argv[1]
    with open(ruta_entrada, 'r', encoding='utf-8') as f:
        entrada = f.read()

    errores_lexicos.clear()
    js_codigo.clear()

    parser.parse(entrada, lexer=lexer) # Se ejecuta el parser con el lexer 

    salida = []
    if errores_lexicos:
        salida.extend(errores_lexicos)
    elif any("Error sintáctico" in linea for linea in js_codigo):
        salida = [linea for linea in js_codigo if "Error sintáctico" in linea]
    else:
        salida = js_codigo

    ruta_salida = f"{ruta_entrada}.js"
    with open(ruta_salida, 'w', encoding='utf-8') as f_out:
        for linea in salida:
            print(linea)
            f_out.write(linea + '\n')

if __name__ == '__main__':
    main()
````

### **TestSuite**:
Se implementan **12 casos de prueba automatizados** con `pytest` cubriendo:
* Los **7 eventos compartidos** (en modo Pointer unificado y en modo explícito `con mouse`) combinados con bloques de acciones UI.
* Los **4 eventos exclusivos de Pointer Events** con acciones.
* Los **5 eventos exclusivos de Mouse / UI** con acciones.
* Las **5 acciones de manipulación de UI** (`CAMBIAR`, `MOSTRAR`, `OCULTAR`, `HABILITAR`, `DESHABILITAR`).
* **Errores léxicos** (caracteres inválidos como `@`, `$`).
* **Errores sintácticos** (orden incorrecto de palabras clave o ausencia de delimitador `DP` / acciones).

````python
%%file traductor/tests.py
import subprocess
import sys
from pathlib import Path
import pytest

def ejecutar_traductor(nombre_archivo, contenido):
    ruta_in = Path(f"traductor/{nombre_archivo}")
    ruta_in.write_text(contenido, encoding="utf-8")
    resultado = subprocess.run(
        [sys.executable, 'traductor/main.py', str(ruta_in)],
        capture_output=True,
        text=True
    )
    ruta_out = Path(f"traductor/{nombre_archivo}.js")
    assert resultado.returncode == 0
    assert ruta_out.exists()
    return ruta_out.read_text(encoding="utf-8").strip()

# --- CASOS VÁLIDOS (PROGRAMAS CORRECTOS) ---

def test_01_compartidos_pointer_down_up_move():
    contenido = (
        'cuando presiono sobre #btn : Mostrar "estado"\n'
        'cuando suelto sobre #btn : Ocultar "estado"\n'
        'cuando me muevo sobre .area : Cambiar "cursor" Opacidad "0.8"'
    )
    salida = ejecutar_traductor("test01.txt", contenido)
    assert "pointerdown" in salida
    assert "pointerup" in salida
    assert "pointermove" in salida
    assert "estado" in salida

def test_02_compartidos_pointer_over_out_enter_leave():
    contenido = (
        'cuando estoy sobre #caja : Mostrar "tooltip"\n'
        'cuando no estoy sobre #caja : Ocultar "tooltip"\n'
        'cuando ingreso con pointer sobre .panel : Habilitar "panel"\n'
        'cuando salgo sobre .panel : Deshabilitar "panel"'
    )
    salida = ejecutar_traductor("test02.txt", contenido)
    assert "pointerover" in salida
    assert "pointerout" in salida
    assert "pointerenter" in salida
    assert "pointerleave" in salida

def test_03_compartidos_modo_mouse_explicito():
    contenido = (
        'cuando presiono con mouse sobre #btn : Mostrar "alerta" Cambiar "alerta" Texto "Activo"\n'
        'cuando me muevo con mouse sobre #canvas : Cambiar "coordenadas" Color "azul"'
    )
    salida = ejecutar_traductor("test03.txt", contenido)
    assert "mousedown" in salida
    assert "mousemove" in salida

def test_04_exclusivos_pointer_captura_y_cancelacion():
    contenido = (
        'cuando cancelo puntero sobre #slider : Deshabilitar "barra"\n'
        'cuando capturo puntero sobre #slider : Habilitar "barra"\n'
        'cuando pierdo captura sobre #slider : Cambiar "barra" Fondo "gris"'
    )
    salida = ejecutar_traductor("test04.txt", contenido)
    assert "pointercancel" in salida
    assert "gotpointercapture" in salida
    assert "lostpointercapture" in salida

def test_05_exclusivo_pointer_raw():
    salida = ejecutar_traductor(
        "test05.txt",
        'cuando detecto puntero sobre #pad-dibujo : Mostrar "trazo"'
    )
    assert "pointerrawupdate" in salida

def test_06_exclusivos_mouse_clics_y_acciones_multiples():
    contenido = (
        'cuando hago clic sobre #btn-enviar :\n'
        '    Cambiar "resultado" Texto "Correcto"\n'
        '    Cambiar "resultado" Color "verde"\n'
        '    Mostrar "resultado"\n'
        '    Ocultar "mensaje"\n'
        '    Deshabilitar "boton"\n'
        'cuando doble clic sobre .item : Mostrar "detalle"\n'
        'cuando clic auxiliar sobre #tab : Ocultar "pestana"'
    )
    salida = ejecutar_traductor("test06.txt", contenido)
    assert "click" in salida
    assert "dblclick" in salida
    assert "auxclick" in salida
    assert "Correcto" in salida

def test_07_exclusivos_mouse_rueda_y_menu_contextual():
    contenido = (
        'cuando giro rueda sobre #mapa : Cambiar "zoom" Tamaño "18px"\n'
        'cuando menu contextual sobre .fila : Mostrar "menu-opciones"'
    )
    salida = ejecutar_traductor("test07.txt", contenido)
    assert "wheel" in salida
    assert "contextmenu" in salida

def test_08_multiples_instrucciones_mixtas(): 
    contenido = (
        'cuando clic sobre #inicio : Mostrar "bienvenida"\n'
        'cuando ingreso con pointer sobre #menu : Habilitar "sub"\n'
        'cuando salgo con mouse sobre #menu : Deshabilitar "sub"'
    )
    salida = ejecutar_traductor("test08.txt", contenido)
    assert salida.count("addEventListener") == 3

# --- CASOS DE ERROR LÉXICO ---

def test_09_error_lexico_caracter_ilegal():
    salida = ejecutar_traductor(
        "test09.txt",
        'cuando presiono sobre @boton_invalido : Mostrar "resultado"'
    )
    assert "Error léxico" in salida

def test_10_error_lexico_simbolo_moneda():
    salida = ejecutar_traductor(
        "test10.txt",
        'cuando clic $ sobre #precio : Ocultar "total"'
    )
    assert "Error léxico" in salida

# --- CASOS DE ERROR SINTÁCTICO ---

def test_11_error_sintactico_orden_invertido():
    salida = ejecutar_traductor(
        "test11.txt",
        'presiono cuando sobre #btn : Mostrar "resultado"'
    )
    assert "Error sintáctico" in salida

def test_12_error_sintactico_falta_dp_o_acciones():
    salida = ejecutar_traductor(
        "test12.txt",
        'cuando doble clic sobre #btn'
    )
    assert "Error sintáctico" in salida
````

* La arquitectura propuesta permite aprovechar lo mejor de ambos paradigmas: **PLY (`lex` y `yacc`)** garantiza una validación estricta, determinística y rápida de la estructura gramatical del lenguaje fuente (incluyendo tanto los eventos disparadores como el delimitador de bloque `DP` y las acciones UI encadenadas), rechazando entradas mal formadas antes de consumir recursos de inferencia.
* La integración con **Groq (`gpt-oss-20b`)** simplifica la traducción de construcciones declarativas en español hacia código JavaScript estándar, resolviendo de manera limpia la dualidad entre los **16 eventos de interacción** (Pointer y mouse) como las operaciones de manipulación del DOM (`CAMBIAR`, `MOSTRAR`, `OCULTAR`, `HABILITAR`, `DESHABILITAR`).
