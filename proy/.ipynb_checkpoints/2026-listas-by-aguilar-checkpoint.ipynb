# Listas

## Objetivo

Desarrollar un traductor que gestione listas de elementos y sus instrucciones a código ejecutable en Python.

## Alcance

* Crear una lista

  #### **AGREGACION**:

* Agregar elementos

* Insertar elemento por posicion#

* Unir listas# 

* reemplazar elemento#

  #### **ELIMINACION**:

* Eliminar elemento

* Eliminar elemento por posicion#

* Eliminar ultimo elemento*

* Vaciar Lista#

  #### **VISUALIZACION**:

* Mostrar la lista

* Mostrar posicion en la lista de un elemento#(ver posicion del elemento)

* traer posicion#(mostrar por pocicion)

  #### **ORGANIZACION**:

* Consultar si un elemento existe

* Ordenar elementos

* Invertir lista#

* Contar elementos de lista

* contar cuantos veces se repite elemento de lista




## Especificaciones léxicas

|    Lexema (Palabra clave)    |   Token (Etiqueta)     |       Patron                      |
|--                            |--                      | --                                |
|        lista                 |      LISTA             |       lista                       |
|        agregar	           |     AGREGAR	        |       agregar                     |
|        insertar              |     INSERTAR           |       insertar                    |
|        unir                  |     Unir               |       unir                        |
|        reemplazar            |     REEMPLAZAR         |       reemplazar                  |
|        eliminar	           |     ELIMINAR	        |       eliminar                    |
|        eliminar_posicion     |     ELIMINAR_POS       |       eliminar_posicion           |
|        eliminar_ultimo       |     ELIMINAR_ULT	    |       eliminar_ultimo             |
|        vaciar  	           |     VACIAR 	        |       vaciar                      |
|        mostrar	           |     MOSTRAR	        |       mostrar                     |
|        mostrar_posicion	   |     MOSTRAR_POSICION   |       mostrar_posicion            |
|        traer_posicion	       |     TRAER_POS	        |       traer_posicion              |
|        existe                |     EXISTE  	        |       existe                      |

## Especificaciones sintácticas

```
 <Programa> ::= <Instruccion> | <Instruccion> <Programa>

<Instruccion> ::= <CrearLista> 
                | <Agregar> 
                | <Insertar>
                | <Unir>
                | <Reemplazar>
                | <Eliminar>
                | <EliminarPos>
                | <EliminarUltimo>
                | <Vaciar>
                | <Mostrar>
                | <MostrarPos>
                | <TraerPos>
                | <Existe>
                | <Ordenar>
                | <Invertir>
                | <Tamano>
                | <Repeticiones>

/* CREACIÓN Y AGREGACIÓN */
<CrearLista> ::= LISTA ID
<Agregar>    ::= AGREGAR ID CADENA
<Insertar>   ::= INSERTAR ID NUMERO CADENA
<Unir>       ::= UNIR ID ID

/* ACTUALIZACIÓN */
<Reemplazar> ::= REEMPLAZAR ID NUMERO CADENA

/* ELIMINACIÓN */
<Eliminar>       ::= ELIMINAR ID CADENA
<EliminarPos>    ::= ELIMINAR_POS ID NUMERO
<EliminarUltimo> ::= ELIMINAR_ULTIMO ID
<Vaciar>         ::= VACIAR ID

/* VISUALIZACIÓN */
<Mostrar>    ::= MOSTRAR ID
<MostrarPos> ::= MOSTRAR_POS ID CADENA
<TraerPos>   ::= TRAER_POS ID NUMERO

/* ORGANIZACIÓN Y CONSULTAS */
<Existe>       ::= EXISTE ID CADENA
<Ordenar>      ::= ORDENAR ID
<Invertir>     ::= INVERTIR ID
<Tamano>       ::= TAMANO ID
<Repeticiones> ::= REPETICIONES ID CADENA
```

## Especificaciones semánticas

|              Regla                          |    Acciones Semánticas                                               |
|--                                           |--                                                                    |    
| <Programa> ::= <Instruccion>                |    Devuelve el código de la instrucción.                             |
| <Programa> ::= <Instruccion> <Programa>     |    Une la instrucción con las demás agregando un salto de línea      |  
| <CrearLista> ::= LISTA ID                   |    Traduce a: ID = []                                                |
| <Agregar> ::= AGREGAR ID CADENA             |    Traduce a: ID.append(CADENA)                                      |
| <Insertar> ::= INSERTAR ID NUMERO CADENA    |    Traduce a: ID.insert(NUMERO, CADENA)                              | 
| <Unir> ::= UNIR ID ID                       |    Traduce a: ID1.extend(ID2) (Une la segunda lista a la primera)    |
| <Reemplazar> ::= REEMPLAZAR ID NUMERO CADENA|    Traduce a: ID[NUMERO] = CADENA                                    |
| <Eliminar> ::= ELIMINAR ID CADENA           |    Traduce a: ID.remove(CADENA)                                      |
| <EliminarPos> ::= ELIMINAR_POS ID NUMERO    |    Traduce a: del ID[NUMERO]                                         |
| <EliminarUltimo> ::= ELIMINAR_ULTIMO ID     |    Traduce a: ID.pop()                                               |
| <Vaciar> ::= VACIAR ID                      |    Traduce a: ID.clear()                                             |
| <Mostrar> ::= MOSTRAR ID                    |    Traduce a: print(ID)                                              |
| <MostrarPos> ::= MOSTRAR_POS ID CADENA      |    Traduce a: print(ID.index(CADENA))                                | 
| <TraerPos> ::= TRAER_POS ID NUMERO          |    Traduce a: print(ID[NUMERO])                                      |
| <Existe> ::= EXISTE ID CADENA               |    Traduce a: print(CADENA in ID)                                    |
| <Ordenar> ::= ORDENAR ID                    |    Traduce a: ID.sort()                                              |
| <Invertir> ::= INVERTIR ID                  |    Traduce a: ID.reverse()                                           |
| <Tamano> ::= TAMANO ID                      |    Traduce a: print(len(ID))                                         |
| <Repeticiones> ::= REPETICIONES ID CADENA   |    Traduce a: print(ID.count(CADENA))                                |

                                 

## Implementació

````python
import ply.lex as lex

# ANALIZADOR LÉXICO: 

reservadas = {
    'lista': 'LISTA', 'agregar': 'AGREGAR', 'insertar': 'INSERTAR',
    'unir': 'UNIR', 'reemplazar': 'REEMPLAZAR', 'eliminar': 'ELIMINAR',
    'eliminar_posicion': 'ELIMINAR_POS', 'eliminar_ultimo': 'ELIMINAR_ULTIMO',
    'vaciar': 'VACIAR', 'mostrar': 'MOSTRAR', 'mostrar_posicion': 'MOSTRAR_POS',
    'traer_posicion': 'TRAER_POS', 'existe': 'EXISTE', 'ordenar': 'ORDENAR',
    'invertir': 'INVERTIR', 'tamano': 'TAMANO', 'repeticiones': 'REPETICIONES'
}

tokens = ['ID', 'CADENA', 'NUMERO'] + list(reservadas.values())
t_ignore = ' \t'

def t_ID(t):
    r'[a-zA-Z][a-zA-Z0-9_]*'
    t.type = reservadas.get(t.value, 'ID')
    return t

def t_CADENA(t):
    r'\"([^\\\n]|(\\.))*?\"'
    return t

def t_NUMERO(t):
    r'[0-9]+'
    return t

def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

def t_error(t):
    print(f"Error léxico en '{t.value[0]}' (Línea {t.lexer.lineno})")
    t.lexer.skip(1)

lexer = lex.lex()
````

````python
import ollama
# analisis sintactico (LLAMA 3.1) 

def motor_semantico_ia(instruccion_cruda):
    """
    Motor Semántico con el diccionario completo de 17 instrucciones.
    """
    prompt_sistema = """Eres un traductor estricto a Python 3. 
Traduce el 'Input' al 'Output' usando EXACTAMENTE esta tabla de patrones:

lista [ID]  =>  [ID] = []
agregar [ID] [CADENA]  =>  [ID].append([CADENA])
insertar [ID] [NUMERO] [CADENA]  =>  [ID].insert([NUMERO], [CADENA])
unir [ID1] [ID2]  =>  [ID1].extend([ID2])
reemplazar [ID] [NUMERO] [CADENA]  =>  [ID][NUMERO] = [CADENA]
eliminar [ID] [CADENA]  =>  [ID].remove([CADENA])
eliminar_posicion [ID] [NUMERO]  =>  del [ID][NUMERO]
eliminar_ultimo [ID]  =>  [ID].pop()
vaciar [ID]  =>  [ID].clear()
mostrar [ID]  =>  print([ID])
mostrar_posicion [ID] [CADENA]  =>  print([ID].index([CADENA]))
traer_posicion [ID] [NUMERO]  =>  print([ID][NUMERO])
existe [ID] [CADENA]  =>  print([CADENA] in [ID])
ordenar [ID]  =>  [ID].sort()
invertir [ID]  =>  [ID].reverse()
tamano [ID]  =>  print(len([ID]))
repeticiones [ID] [CADENA]  =>  print([ID].count([CADENA]))

REGLAS ABSOLUTAS:
1. Responde SOLO con el código final a la derecha de la flecha (=>).
2. NUNCA inventes líneas de código ni reinicies variables (como 'ID = []') a menos que el Input sea explícitamente el comando 'lista'.
3. Prohibido usar markdown, comillas invertidas, o dar explicaciones.

Ejemplos de comportamiento esperado:
Input: lista repuestos
Output: repuestos = []

Input: agregar repuestos "tuerca"
Output: repuestos.append("tuerca")

Input: repeticiones repuestos "tuerca"
Output: print(repuestos.count("tuerca"))

Input: unir stock_a stock_b
Output: stock_a.extend(stock_b)

Input: mostrar_posicion stock_a "tuerca"
Output: print(stock_a.index("tuerca"))

Input: traer_posicion stock_a 2
Output: print(stock_a[2])

Input: invertir invitados
Output: invitados.reverse()

"""

    mensajes = [
        {'role': 'system', 'content': prompt_sistema},
        {'role': 'user', 'content': f"Input: {instruccion_cruda}\nOutput:"}
    ]
    
    try:
        respuesta = ollama.chat(model='llama3.1', messages=mensajes, options={"temperature": 0.0})
        
        codigo_generado = respuesta['message']['content'].strip()
        
        # Limpieza única de markdown y backticks
        codigo_generado = codigo_generado.replace("```python", "").replace("```", "").replace("`", "").strip()
        
        # Parche de seguridad por si la IA devuelve el comando original hace "eco"
        if codigo_generado.startswith(instruccion_cruda.split()[0]):
            return f"# Fallo de IA en: {instruccion_cruda}"
            
        return codigo_generado
    except Exception as e:
        return f"# Error de IA: {e}"
````

````python
import ply.yacc as yacc

# concatenación
def p_programa_multiple(p):
    '''programa : instruccion programa'''
    p[0] = p[1] + '\n' + p[2]

def p_programa_simple(p):
    '''programa : instruccion'''
    p[0] = p[1]

# semantica (instrucciones)

def p_instruccion_crear(p):
    '''instruccion : LISTA ID'''
    p[0] = motor_semantico_ia(f"lista {p[2]}")

def p_instruccion_agregar(p):
    '''instruccion : AGREGAR ID CADENA'''
    p[0] = motor_semantico_ia(f"agregar {p[2]} {p[3]}")

def p_instruccion_insertar(p):
    '''instruccion : INSERTAR ID NUMERO CADENA'''
    p[0] = motor_semantico_ia(f"insertar {p[2]} {p[3]} {p[4]}")

def p_instruccion_unir(p):
    '''instruccion : UNIR ID ID'''
    p[0] = motor_semantico_ia(f"unir {p[2]} {p[3]}")

def p_instruccion_reemplazar(p):
    '''instruccion : REEMPLAZAR ID NUMERO CADENA'''
    p[0] = motor_semantico_ia(f"reemplazar {p[2]} {p[3]} {p[4]}")

def p_instruccion_eliminar(p):
    '''instruccion : ELIMINAR ID CADENA'''
    p[0] = motor_semantico_ia(f"eliminar {p[2]} {p[3]}")

def p_instruccion_eliminar_pos(p):
    '''instruccion : ELIMINAR_POS ID NUMERO'''
    p[0] = motor_semantico_ia(f"eliminar_posicion {p[2]} {p[3]}")

def p_instruccion_eliminar_ult(p):
    '''instruccion : ELIMINAR_ULTIMO ID'''
    p[0] = motor_semantico_ia(f"eliminar_ultimo {p[2]}")

def p_instruccion_vaciar(p):
    '''instruccion : VACIAR ID'''
    p[0] = motor_semantico_ia(f"vaciar {p[2]}")

def p_instruccion_mostrar(p):
    '''instruccion : MOSTRAR ID'''
    p[0] = motor_semantico_ia(f"mostrar {p[2]}")

def p_instruccion_mostrar_pos(p):
    '''instruccion : MOSTRAR_POS ID CADENA'''
    p[0] = motor_semantico_ia(f"mostrar_posicion {p[2]} {p[3]}")

def p_instruccion_traer_pos(p):
    '''instruccion : TRAER_POS ID NUMERO'''
    p[0] = motor_semantico_ia(f"traer_posicion {p[2]} {p[3]}")

def p_instruccion_existe(p):
    '''instruccion : EXISTE ID CADENA'''
    p[0] = motor_semantico_ia(f"existe {p[2]} {p[3]}")

def p_instruccion_ordenar(p):
    '''instruccion : ORDENAR ID'''
    p[0] = motor_semantico_ia(f"ordenar {p[2]}")

def p_instruccion_invertir(p):
    '''instruccion : INVERTIR ID'''
    p[0] = motor_semantico_ia(f"invertir {p[2]}")

def p_instruccion_tamano(p):
    '''instruccion : TAMANO ID'''
    p[0] = motor_semantico_ia(f"tamano {p[2]}")

def p_instruccion_repeticiones(p):
    '''instruccion : REPETICIONES ID CADENA'''
    p[0] = motor_semantico_ia(f"repeticiones {p[2]} {p[3]}")

def p_error(p):
    if p:
        print(f"Error de sintaxis en '{p.value}'")
    else:
        print("Error de sintaxis al final del archivo")

parser = yacc.yacc(write_tables=False, debug=False)
````
