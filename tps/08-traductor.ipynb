## Ejemplo 1: Mini markdown

* **Identificación**: mmm-2026-by-pandolfo
* **Objetivo**: Transformar programa escrito en lenguaje mmm en código HTML
* **Especificaciones Léxicas**:

| Lexema | Token | Patrón  |
| --     | --    | --      |
| #      | HA    | #       |
| *      | IT    | *       |
| texto  | TXT   | (L\|D)* |

* **Especificaciones Sintácticas**:

```
<Programa> ::= <Linea>
<Programa> ::= <Linea> <Programa>
<Linea> ::= HA TXT
<Linea> ::= HA HA TXT
<Linea> ::= IT TXT
```
* **Especificaciones Semánticas**:

| Regla                                | Acciones Semánticas  |
| --                                   | --                   |
| \<Programa> ::= \<Linea>             | Una línea            |
| \<Programa> ::= \<Linea> \<Programa> | Muchas líneas        |
| \<Linea> ::= HA TXT                  | Trad: \<h1>TXT\<\h1> |
| \<Linea> ::= HA HA TXT               | Trad: \<h2>TXT\<\h2> |
| \<Linea> ::= IT TXT                  | Trad: \<li>TXT\<\li> |

* **Ejemplo**:
    * **Programa Fuente**:
 
        ```
        # UNO
        ## Tecn Web
        * PW1
        * LP
        * BD
        ```

    * **Programa Destino**:
 
        ```
        <html>
        <body>
        <h1>UNO</h1>
        <h2>Tecn Web</h2>
        <ul>
        	<li>PW1</l1>
        	<li>LP</l1>
        	<li>BD</l1>
        </body>
        </html>
        ```
