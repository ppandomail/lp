5. **[3 puntos]** Represéntese en Prolog información sobre estudiantes, las materias que estudian y las materias que han aprobado. A partir de los hechos, defínase las siguientes reglas:
* **companero(X, Y)** que determine si dos estudiantes cursan la misma materia.
* **pendiente(X, Materia)** que determine si un estudiante cursa una materia pero todavía no la ha aprobado.
* **al_dia(X)** que determine si un estudiante ha aprobado la materia que estudia.

Realizar consultas para comprobar el funcionamiento de las reglas.

```python
import janus_swi as janus
janus.consult('parcial', '''
estudia(juan, programacion).
estudia(ana, programacion).
estudia(pedro, matematica).
estudia(laura, ingles).
aprobo(juan, programacion).
aprobo(pedro, matematica).
companero(X, Y) :- estudia(X, Materia), estudia(Y, Materia), X \\= Y.
pendiente(X, Materia) :- estudia(X, Materia), \\+ aprobo(X, Materia).
al_dia(X) :- estudia(X, Materia), aprobo(X, Materia).
''')
# \\+: negación por fallo
print(list(janus.query('companero(juan,ana).')))
print(list(janus.query('pendiente(X,Materia).')))
print(list(janus.query('al_dia(X).')))
```
