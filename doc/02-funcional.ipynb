# Paradigma Funcional
---

## Concepto

* Se basa en tratar el cómputo como la evaluación de funciones matemáticas, evitando el cambio de estado y los datos mutables
* Aunque Python es un lenguaje multiparadigma (donde predomina la orientación a objetos), incluye soporte nativo y bibliotecas para aplicar programación funcional de forma limpia y expresiva

## Características

* Es un paradigma declarativo que se basa en un modelo matemático de composición de funciones
* En este modelo, el resultado de un cálculo es la entrada del siguiente, y así sucesivamente hasta que una composición produce el
valor deseado, tal como sucede en la composición de funciones matemáticas
* Los programas funcionales expresan mejor qué hay que calcular, y no detallan tanto cómo realizar dicho cálculo a diferencia de lo
que ocurre en los programas imperativos
* La característica principal de la programación funcional es que los cálculos se ven como una función matemática que hacen
corresponder entradas y salidas

## Principios fundamentales en Python

* **Funciones de primera clase y de orden superior**: las funciones son objetos. Se puede asignar a variables, pasarlas como argumentos y retornarlas desde otras funciones
* **Inmutabilidad**: en lugar de modificar estructuras existentes, se crean nuevas variables o estructuras de datos. En Python se logra mediante colecciones inmutables (tuple, frozenset) o definiendo dataclasses(frozen=True)
* **Evitar efectos secundarios (Funciones Puras)**: una función pura siempre devuelve el mismo resultado para los mismos argumentos y no modifica el entorno global ni el estado externo

## Lenguajes

### Puros

* En estos lenguajes **no existen las variables mutables ni los efectos secundarios**
* Todo se resuelve mediante funciones puras y evaluación matemática
* Entre los que se encuentran:
    * Haskell
    * Miranda
    * Gofer

### Nativos (Impuros / Híbridos)

* Fueron creados desde cero bajo el paradigma funcional, pero **permiten modificar datos o ejecutar efectos secundarios** de forma explícita por razones de rendimiento o practicidad
* Existen varios lenguajes con esta clasificación, algunos son:
    * Lisp
    * Scheme
    * Scala
    * Clojure
    * ML
    * Ocam
    * Racket

### Multiparadigma (Adaptados)

* Nacieron bajo paradigmas imperativos u orientados a objetos, pero **agregaron características funcionales** en versiones posteriores
* Los más conocidos son:
    * Python
    * JavaScript
    * Java

## Aplicación

* **Demostraciones de teoremas**: por su naturaleza "funcional" este paradigma es útiles en la demostración automática de teoremas, ya que permite especificar de manera adecuada y precisa problemas matemáticos
* **Creación de compiladores, analizadores sintácticos**: su naturaleza inherentemente recursiva le permite modelar adecuadamente estos problemas
* Resolver problemas que requieran **demostraciones por inducciones**
* En la industria se puede usar para resolver **problemas matemáticos complejos**

## Funciones lambda

* Función anónima (sin nombre)
* Se usa para abreviar
* Nunca tendrá condicionales ni bucles

```python
area = lambda b, h : (b * h) / 2
```

```python
area(5, 7)
```

```python
cubo = lambda n : pow(n, 3)
```

```python
cubo(2)
```

## Funciones de orden superior

* Funciones pueden pasarse como argumentos de una función

```python
def aplica(f, arg):
    return f(arg)

def cuadrado(n):
    return n * n

def cubo(n):
    return n**3
```

```python
aplica(cuadrado, 2)
```

```python
aplica(cubo, 2)
```

### filter

* Verifica que los elementos de una secuencia cumplen una condición, devolviendo un iterador con los elementos que cumple dicha condición

```python
def par(n):
  return n % 2 == 0
```

```python
list(filter(par, [17, 24, 7, 39, 8, 51, 92]))
```

### map

* Aplica una función a cada elemento de una lista iterable (listas, tuplas, etc.)  devolviendo una lista con los resultados

```python
def cuadrado(n):
    return n * n
```

```python
list(map(cuadrado, [1, 2, 3, 4, 5]))
```

### reduce

* Operar todos elementos de una colección iterable (reduce)
* reduce(f, l): aplicar la función f a los dos primeros elementos de la secuencia l
* Con ese valor vuelve aplicar con el siguiente

```python
from functools import reduce

def producto(n, m):
    return n * m
```

```python
reduce(producto, [1, 2, 3, 4, 5])
```
